```python
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n-1)
print(factorial(5))
def fib(n):
    if n <= 2:
        return 1
    return fib(n-1) + fib(n-2)
print([fib(i) for i in range(1,11)])
from functools import lru_cache
@lru_cache(maxsize=None)
def fib_fast(n):
    if n <= 2:
        return 1
    return fib_fast(n-1) + fib_fast(n-2)
import time
start = time.time()
print(fib_fast(100))
print(f"耗时：{time.time() - start:.4f}秒")
def recursive_sum(lst):
    if not lst:
        return 0
    return lst[0] + recursive_sum(lst[1:])
print(recursive_sum([1,2,3,4,5]))
def flatten(nested):
    result = []
    for item in nested:
        if isinstance(item,list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result
deep = [1,[2,[3,4],5],[[6]],7]
print(flatten(deep))
def print_tree(structure,indent=0):
    for name,children in structure.items():
        print("  " * indent + "├── " + name)
        if children:
            print_tree(children,indent + 1)
file_tree = {
    "项目": {
        "src": {
            "main.py": {},
            "utils.py": {},
        },
        "tests": {
            "test_main.py": {},
        },
        "README.md": {},
    }
}
print_tree(file_tree)
```

```python
x = "global"
def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print("inner:",x)
    inner()
    print("outer:",x)
outer()
print("global:",x)
count = 0
def increment():
    global count
    count += 1
increment()
increment()
print(count)
def counter():
    count = 0
    def increment():
        nonlocal count
        count += 1
        return count
    return increment
c = counter()
print(c())
print(c())
print(c())
def counter_with_list():
    data = {"count":0}
    def increment():
        data["count"] += 1
        return data["count"]
    return increment
c3 = counter_with_list()
print(c3(),c3(),c3())
```

```python
def make_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier
double = make_multiplier(2)
triple = make_multiplier(3)
print(double(5))
print(triple(5))
print(double(10))
print(double.__closure__)
print(double.__closure__[0].cell_contents)
def create_counter(start=0):
    count = start
    def counter(step=1):
        nonlocal count
        count += step
        return count
    return counter
c1 = create_counter()
c2 = create_counter(100)
print(c1(),c1(),c1())
print(c2(),c2())
print(c1())
```

```python
def power_factory(exp):
    def power(base):
        return base ** exp
    return power
square = power_factory
cube = power_factory(3)
print(square(5))
print(cube(5))
```

```python
def say_hello(name):
    print(f"你好,{name}")
greet = say_hello
greet("小明")
def call_twice(func):
    func("第一次")
    func("第二次")
call_twice(say_hello)
import time
def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        elapsed = time.time() - start
        print(f"[{func.__name__}] 耗时 {elapsed:.4f} 秒")
        return result
    return wrapper
@timer
def slow_func(n):
    total = 0
    for i in  range(n):
        total += i ** 2
    return total
print(slow_func(1000000))

from functools import wraps
def logger(func):
    @wraps(func)
    def wrapper(*args,**kwargs):
        print(f"调用 {func.__name__}，参数：{args}, {kwargs}")
        result = func(*args,**kwargs)
        print(f"{func.__name__} 返回：{result}")
        return result
    return wrapper
@logger
def add(a,b):
    """两数相加"""
    return a + b
print(add(3,5))
print(add.__name__)
print(add.__doc__)
def repeat(times=2):
    def decorator(func):
        @wraps(func)
        def wrapper(*args,**kwargs):
            results = []
            for i in range(times):
                print(f"  第 {i+1}/{times} 次执行")
                results.append(func(*args,**kwargs))
            return results
        return wrapper
    return decorator
@repeat(3)
def greet(name):
    print(f"  你好，{name}!")
    return f"greeted {name}"
greet("小明")
@timer
@logger
def compute(n):
    return sum(i ** 2 for i in range(n))
compute(100000)
```

```python
import time
from functools import lru_cache,wraps
print("=" * 50)
print("练习1：lambda 实战")
print("=" * 50)
students = [
    {"name": "小明", "chinese": 85, "math": 92, "english": 78},
    {"name": "小红", "chinese": 90, "math": 88, "english": 95},
    {"name": "小刚", "chinese": 72, "math": 65, "english": 80},
    {"name": "小丽", "chinese": 95, "math": 97, "english": 92},
]
for s in students:
    s["total"] = s['chinese'] + s["math"] + s["english"]
ranked = sorted(students,key=lambda s:s["total"],reverse=True)
print("\n成绩排名：")
for i,s in enumerate(ranked,1):
    print(f"  {i}. {s['name']}：总分{s['total']}")
best_math = max(students,key=lambda s:s["math"])
print(f"\n数学最高：{best_math['name']}（{best_math['math']}分）")
top_students = list(filter(lambda s:s["total"] > 260,students))
print(f"总分>260：{[s['name'] for s in top_students]}")
```

```python
print("\n" + "=" * 50)
print("练习2：递归实战")
print("=" * 50)
def digit_sum(n):
    n = abs(n)
    if n < 10:
        return n
    return n % 10 + digit_sum(n // 10)
print(f"\ndigit_sum(12345) = {digit_sum(12345)}")
def gcd(a,b):
    return a if b == 0 else gcd(b,a % b)
print(f"gcd(48,18) = {gcd(48,18)}")
print(f"gcd(1071,462) = {gcd(1071,462)}")
```
