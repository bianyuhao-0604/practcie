```python
scores = [
    [85,92,78],
    [90,88,95],
    [72,65,80],
    [95,97,92],
]
print(scores[0])
print(scores[0][1])
print(scores[2][0])
scores[2][1] = 70
print("修改后:",scores[2])
print("\n=== 成绩表 ===")
subjects = ["语文","数学","英语"]
for i,student in enumerate(scores):
    print(f"学生{i+1}:",end="")
    for j,score in enumerate(student):
        print(f"{subjects[j]}={score}",end=" ")
    print()
print("\n=== 各科统计 ===")
for j in range(len(subjects)):
    column = [scores[i][j] for i in range(len(scores))]
    avg = sum(column) / len(column) 
    print(f"{subjects[j]}:最高{max(column)},最低{min(column)},平均{avg:.1f}")
```

```python
print("=== 九九乘法表（基础版） ===")
for i in range(1,10):
    for j in range(1,i+1):
        print(f"{j}x{i}={i*j}",end="\t")
    print()
print()
print("=== 九九乘法表（美化版） ===")
print("    ",end="")
for j in range(1,10):
    print(f"    {j}    ",end="")
print()
print("    " + "-" * 63)
for i in range(1,10):
    print(f"{i} |",end="")
    for j in range(1,10):
        product = i * j
        if j < i:
            print(f"      ",end="")
        else:
            print(f"   {product:3d}   ",end="")
    print()
print()
```

```python
for i in range(1,10):
    for j in range(1,i+1):
        print(f"{j}x{i}={i*j}",end="\t")
    print()
students = [
    ["小明", 85, 92, 78],
    ["小红", 90, 88, 95],
    ["小刚", 72, 65, 80],
    ["小丽", 95, 97, 92],
    ["小华", 60, 55, 70],
]
print("\n=== 学生成绩单 ===")
print(f"{'姓名':<6}{'语文':>6}{'数学':>6}{'英语':>6}{'总分':>6}{'平均分':>8}")
print("-" * 40)
for student in students:
    name = student[0]
    scores = student[1:]
    total = sum(scores)
    avg = total / len(scores)
    print(f"{name:<6}{scores[0]:>6}{scores[1]:>6}{scores[2]:>6}{total:>6}{avg:>8.1f}")
print("\n--- 排名 ---")
students_sorted = sorted(students,key=lambda s: sum(s[1:]),reverse=True)
for rank,student in enumerate(students_sorted,start=1):
    total = sum(student[1:])
    print(f"第{rank}名：{student[0]}，总分{total}") 
```

```python
matrix = [
    [1,2,3],
    [4,5,6]
]
print("\n=== 矩阵转置 ===")
print("原始矩阵：")
for row in matrix:
    print(row)
rows = len(matrix)
cols = len(matrix[0])
transposed = []
for j in range(cols):
    new_row = []
    for i in range(rows):
        new_row.append(matrix[i][j])
    transposed.append(new_row)
print("转置后：")
for row in transposed:
    print(row)
```

```python
print("\n=== 杨辉三角 ===")
n = 10
triangle = []
for i in range(n):
    row = [1]
    if i > 0:
        prey_row = triangle[i - 1]
        for j in range(len(prey_row)-1):
            row.append(prey_row[j]+prey_row[j+1])
        row.append(1)
    triangle.append(row)
width = n * 4
for row in triangle:
    line = "  ".join(f"{num:2d}"for num in row)
    print(line.center(width))
```

```python
print("\n=== 购物车模拟 ===")
cart = []
cart.append(["Python入门书", 59.9, 1])
cart.append(["机械键盘", 299.0, 1])
cart.append(["USB数据线", 19.9, 3])
cart.append(["鼠标垫", 29.9, 2])
print(f"{'商品名称':<14}{'单价':>8}{'数量':>6}{'小计':>10}")
print("-" * 42)
total_price = 0
for item in cart:
    name,price,qty = item
    subtotal = price * qty
    total_price += subtotal
    print(f"{name:<14}{price:>8.1f}{qty:>6}{subtotal:>10.1f}")
print("-" * 42)
print(f"{'总计':<28}{total_price:>10.1f}")
most_expensive = max(cart,key=lambda x: x[1])
cart.remove(most_expensive)
print(f"\n已移除最贵商品:{most_expensive[0]}(￥{most_expensive[1]})")
total_price = sum(item[1] * item[2] for item in cart)
print(f"移除后总价:￥{total_price:.1f}")
```
