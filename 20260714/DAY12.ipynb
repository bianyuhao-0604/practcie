```python
class Animal:
    def __init__(self,name,sound):
        self.name = name
        self.sound = sound
    def speak(self):
        return f'speaking'
class Dog(Animal):
    def __init__(self,name):
        super().__init__(name,'wangwang')
    def speak(self):
        return f'woof'
class Cat(Animal):
    def __init__(self,name):
        super().__init__(name,'miaomiao')
    def speak(self):
        return f'miaomiao'
dog1 = Dog('bob')
print(dog1.speak())
        
```

```python
import math
class Shape:
    def area(self):
        raise NotImplemented("子类必须实现 area() 方法！")
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return math.pi * (self.radius ** 2)
class Square(Shape):
    def __init__(self,side):
        self.side = side
    def area(self):
        return self.side ** 2
circle = Circle(1)
print(circle.area())
square = Square(2)
print(square.area())

shapes = [Circle(1),Circle(3),Square(7),Circle(2.5)]
print("--- 图形面积报告 ---")
for i in shapes:
    print(int(i.area()))
```

```python
class Vehicle:
    def __init__(self,brand,max_speed):
        self.brand = brand
        self.max_speed =max_speed
    
    def info(self):
        return f'[{self.brand}] 最高时速: {self.max_speed}km/h'

class Car(Vehicle):
    def __init__(self,brand,max_speed,doors):
        super().__init__(brand,max_speed)
        self.doors = doors
    def info(self):
        return f'{super().info()} | {self.doors}'
car1 = Car('奥迪',200,8)
print(car1.info())

class ElectricCar(Vehicle):
    def __init__(self,brand,max_speed,battery_kwh):
        super().__init__(brand,max_speed)
        self.battery_kwh = battery_kwh
    def info(self):
        return f'{super().info()} | {self.battery_kwh}'
electriccar = ElectricCar('雅迪',80,2000)
print(electriccar.info())
    
```

```python
class Fruit:
    def __init__(self,name,price):
        self.name = name
        self.price = price

class Apple(Fruit):pass
class Banana(Fruit):pass
class Orange(Fruit):pass

def apply_discount(fruit:Fruit):
    if isinstance(fruit,Apple):
        discount = 0.8
    elif isinstance(fruit,Banana):
        discount = 0.7
    else:
        discount = 0.9
    
    final_price = fruit.price * discount
    print(f"{fruit.name} 原价: ¥{fruit.price} -> 折后价: ¥{final_price:.2f}")

fruits = [Apple("红富士",10),Banana("海南蕉",6),Orange("赣南脐橙",8)]
for f in fruits:
    apply_discount(f)

print("\n--- 类关系检查 ---")
print(f"Apple 是 Fruit 的子类吗？ {issubclass(Apple, Fruit)}")   
print(f"Apple 是 Banana 的子类吗？ {issubclass(Apple, Banana)}")
```

```python
class Employee:
    def __init__(self,name,base_salary):
        self.name = name
        self.base_salary = base_salary
    def calculate_salary(self):
        return self.base_salary
    def __str__(self):
        return f"{self.name} 的本月薪资: ¥{self.calculate_salary():,.2f}"
class Manager(Employee):
    def __init__(self,name,base_salary,bonus):
        super().__init__(name,base_salary)
        self.bonus = bonus
    def calculate_salary(self):
        return self.base_salary + self.bonus
class Developer(Employee):
    def __init__(self,name,base_salary,overtime_hours):
        super().__init__(name,base_salary)
        self.overtime_hours = overtime_hours
    def calculate_salary(self):
        return self.base_salary + (self.overtime_hours * 150)

staff = [
    Manager('王总',25000,10000),
    Developer('小李',15000,30),
    Employee('前台小张',6000)
]
for i in staff:
    print(i)
```

```python
class BaseNotifier:
    def send(self, message):
        raise NotImplementedError

class EmailNotifier(BaseNotifier):
    def __init__(self, email_addr):
        self.email_addr = email_addr
    def send(self, message):
        print(f"[邮件] 发送至 {self.email_addr}: {message}")

class SMSNotifier(BaseNotifier):
    def __init__(self, phone):
        self.phone = phone
    def send(self, message):
        print(f"[短信] 发送至 {self.phone}: {message}")

class WeChatNotifier(BaseNotifier):
    def send(self, message):
        print(f"[微信] 推送至服务号: {message}")

class NotificationManager:
    def __init__(self):
        self.channels = []
        
    def add_channel(self, notifier: BaseNotifier):
        self.channels.append(notifier)
        
    def broadcast(self, message):
        print(f"=== 开始广播: {message} ===")
        for channel in self.channels:
            channel.send(message)  # 多态调用

# 测试
manager = NotificationManager()
manager.add_channel(EmailNotifier("admin@company.com"))
manager.add_channel(SMSNotifier("13800000000"))
manager.add_channel(WeChatNotifier())

manager.broadcast("服务器CPU使用率超过90%，请及时处理！")
```

```python
class Engine:
    def __init__(self,horsepower):
        self.horsepower = horsepower
        self.is_running = False
    def start(self):
        self.is_running = True
        print(f'发动机已运行,马力{self.horsepower}匹。')
    def stop(self):
        self.is_running = False
        print(f'发动机已熄火')
class Car:
    def __init__(self,model,engine:Engine):
        self.model = model
        self.engine = engine
    def drive(self):
        if not self.engine.is_running:
            self.engine.start()
            print(f'{self.model}正在行驶')
v8_engine = Engine(1000)
car = Car('大众',v8_engine)
car.drive()

v4_engine = Engine(500)
print(f'升级后马力:{v4_engine.horsepower}')
    
```

```python
class A:
    def who_am_i(self):
        print('I am A')
class B(A):
    def who_am_i(self):
        print('I am B')
        super().who_am_i()
class C(A):
    def who_am_i(self):
        print('I am C')
        super().who_am_i()
class D(B,C):
    def who_am_i(self):
        print('I am D')
        super().who_am_i()
d = D()
d.who_am_i()
```
