```python
def calculate_bmi(weight,height):
    """
    计算BMI（身体质量指数）

    参数：
        weight: 体重（公斤）
        height: 身高（米）

    返回：
        BMI值（浮点数）

    示例：
        >>> calculate_bmi(70, 1.75)
        22.86
    """
    return weight / (height ** 2)
print(calculate_bmi.__doc__)
bmi = calculate_bmi(70,1.75)
print(f"BMI = {bmi:.2f}")
help(calculate_bmi)
```

```python
def celsius_to_fahrenheit(celsius):
    """摄氏度转华氏度"""
    return celsius * 9 / 5 + 32
def fahrenheit_to_celsius(fahrenheit):
    """华氏度转摄氏度"""
    return (fahrenheit - 32) * 5 / 9
temps_c = [0,20,37,100]
print("===温度转换===")
for c in temps_c:
    f = celsius_to_fahrenheit(c)
    print(f"  {c}°C = {f}°F")
print()
for f in [32,68,98.6,212]:
    c = fahrenheit_to_celsius(f)
    print(f"  {f}°F = {c:.1f}°C")
```

```python
def draw_line(length=20,char="-"):
    print(char * length)
def draw_box(width=20,height=5,char="*"):
    print(char * width)
    for _ in range(height - 2):
        print(char + " " * (width -2 ) + char)
    print(char * width)
print("\n=== 画线 ===")
draw_line()
draw_line(30)
draw_line(15,"=")
print("\n=== 画矩形 ===")
draw_box()
print()
draw_box(25,4,"#")
def average(*numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)
print("\n=== 平均值 ===")
print(f"average(1,2,3)={average(1,2,3):.1f}")
print(f"average(10,20,30,40)={average(10,20,30,40):.1f}")
print(f"average(100)={average(100):.1f}")
def build_profile(**kwargs):
    print("\n=== 个人档案 ===")
    for key,value in kwargs.items():
        print(f"  {key}: {value}")
build_profile(name="小明",age=20,city="苏州",hobby="编程")
build_profile(name="小红",occupation="工程师")
```

```python
def is_palindrome(text):
    cleaned = text.lower().replace(" ","")
    return cleaned == cleaned[::-1]
def count_vowels(text):
    vowels = set("aeiouAEIOU")
    return sum(1 for char in text if char in vowels)
def find_longest_word(text):
    words = text.split()
    if not words:
        return ""
    return max(words,key=len)
print("\n=== 工具函数 ===")
test_strings = ["racecar", "hello", "A man a plan a canal Panama", "Python"]
for s in test_strings:
    print(f"  '{s}' 是回文吗？{is_palindrome(s)}")
sentences = ["Hello World", "Python is great", "AEIOU"]
for s in sentences:
    print(f"  '{s}' 有 {count_vowels(s)} 个元音")
```

```python
def analyze_scores(*scores):
    if not scores:
        return None
    scores_list = list(scores)
    total = sum(scores_list)
    avg = total / len(scores_list)
    passed = sum(1 for s in scores_list if s >= 60)
    grades = {"A":0,"B":0,"C":0,"D":0,"F":0}
    for s in scores_list:
        if s >= 90:grades["A"] += 1
        elif s >= 80:grades["B"] += 1
        elif s >= 70:grades["C"] += 1
        elif s >= 60:grades["D"] += 1
        else:grades["F"] += 1
    return {
        "count": len(scores_list),
        "total": total,
        "average": avg,
        "max": max(scores_list),
        "min": min(scores_list),
        "passed": passed,
        "pass_rate": passed / len(scores_list) * 100,
        "grades": grades,
    }    
print("\n=== 成绩分析 ===")
result = analyze_scores(85, 92, 78, 55, 63, 91, 45, 88, 72, 96)
print(f"  总人数：{result['count']}")
print(f"  总分：{result['total']}")
print(f"  平均分：{result['average']:.1f}")
print(f"  最高分：{result['max']}")
print(f"  最低分：{result['min']}")
print(f"  及格人数：{result['passed']}")
print(f"  及格率：{result['pass_rate']:.0f}%")
print(f"  等级分布：")
for grade, count in result["grades"].items():
    bar = "█" * count
    print(f"    {grade}：{count}人  {bar}")
```
