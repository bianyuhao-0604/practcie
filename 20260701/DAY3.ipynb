```python
def BMI(weight:float,height:float) -> float:
    if height <= 0:
        raise ValueError('')
    bmi = weight / (height ** 2)#防御性编程
    return round(bmi, 2)
print(BMI(70, 1.75))
help(BMI)
print(BMI.__doc__)
```

```python
def log_event(event_name:str,*tags:str,**metadata):
    print(f'Event: {event_name}')
    if tags:
        print(f"Tags: {','.join(tags)}")
    if metadata:
        for k,v in metadata.items():
            print(f'{k}:{v}')
log_event(
    'userlogin',
    'auth','web',
    user_id=101,ip='192.168.1.1'
)


```

```python
#
def analyze_text(text:str) -> tuple[int,int,int]:
    char_count = len(text)
    word_count = len(text.split())#
    line_count = len(text.splitlines())#
    return char_count,word_count,line_count#隐式返回元组
text_data = """hello world
this is a test
day 3
"""
chars,words,lines = analyze_text(text_data)
print(f"字符: {chars}, 单词: {words}, 行数: {lines}")
_,w,_ = analyze_text(text_data)
print(f"单词: {w}")
```

```python
def make_counter():
    count = 0
    def counter():
        nonlocal count
        count += 1
        return count
    return counter
my_counter = make_counter()
print(my_counter())
print(my_counter())
print(my_counter())
```

```python
def multiplier(factor: int):
    # 这个内部函数就是一个闭包
    def multiply(number: int) -> int:
        return number * factor  # factor 来自外部作用域
    return multiply

# 创建特定的乘法器
double = multiplier(2)
triple = multiplier(3)

print(double(5))  # 10
print(triple(5))  # 15

# 查看闭包记住的变量
print(double.__closure__[0].cell_contents)  # 2
print(triple.__closure__[0].cell_contents)  # 3
```

```python
def fib(n:int,_cache:dict={}) -> int:
    if n in _cache:
        return _cache[n]
    if n < 2:
        result = n
    else:
        result = fib(n-1) + fib(n-2)
    _cache[n] = result
    return result
print(fib(10))  
print(fib(50))
```

```python
def remove_duplicates(items):
    seen = set()              # 用集合提升查找效率
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

data = [1, 2, 2, 3, 4, 3, 5]
print(f"去重后: {remove_duplicates(data)}")  # [1, 2, 3, 4, 5]
```
