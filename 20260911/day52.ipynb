```python
import numpy as np
from sklearn.impute import SimpleImputer
X = np.array([[1,2],[np.nan,3],[7,6],[4,np.nan]])
imputer = SimpleImputer(strategy='mean')
X_filled = imputer.fit_transform(X)
print(X_filled)
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
X = np.array([[1,10],[2,11],[3,12],[100,np.nan],[np.nan,13]])
print(X)
mean_imp = SimpleImputer(strategy='mean')
print("均值填充结果:\n",mean_imp.fit_transform(X))
median_imp = SimpleImputer(strategy='median')
print('中位数填充结果:\n',median_imp.fit_transform(X))
```

```python
from sklearn.impute import KNNImputer
X = np.array([[1,2,np.nan],[3,4,5],[np.nan,6,7],[8,9,10]])
print(X)
knn_imp = KNNImputer(n_neighbors=2)
print(knn_imp.fit_transform(X))
```

```python
from sklearn.preprocessing import StandardScaler
X = np.array([[1,100],[2,200],[3,300],[4,400]])
print(X)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("标准化后数据：\n", X_scaled)
print("每列均值：", X_scaled.mean(axis=0))  
print("每列标准差：", X_scaled.std(axis=0))
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np
X = np.array([[10,200],[20,400],[30,600]])
print(X)
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
print(X_scaled.min(axis=0))
print(X_scaled.max(axis=0))
```

```python
from sklearn.preprocessing import LabelEncoder
import numpy as np
y = ['low','medium','high','low','high']
le = LabelEncoder()
y_encoded = le.fit_transform(y)
print("编码结果:",y_encoded)
print("还原结果:",le.inverse_transform(y_encoded))
mapping = {'low':0,'medium':1,'high':2}
y_custom = [mapping[val] for val in y]
print("自定义编码:",y_custom)
```

```python
from sklearn.preprocessing import OneHotEncoder
X_train = np.array([['red'],['green'],['blue']])
print(X_train)
X_test = np.array([['yellow'],['red']])
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoder.fit(X_train)
print('训练集编码:\n',encoder.transform(X_train))
print('测试集编码:\n',encoder.transform(X_test))
```
