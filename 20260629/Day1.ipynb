```python
falsy_values = [False,0,0.0,"",[],{},None,set()]
for val in falsy_values:
    print(f'{str(val):>10} -> bool: {bool(val)}')
truthy_values = [True,1,3.14,"张三",[1],{9}]
for x in truthy_values:
    print(f'{str(x):>10} -> bool: {bool(x)}')
```

```python
s = 'Hello World! 2026'
upper_count = 0
lower_count = 0
for char in s:
    if char.isupper():
        upper_count += 1
    elif char.islower():
        lower_count += 1
print(upper_count,lower_count)

```
