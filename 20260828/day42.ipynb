```python
import tracemalloc
tracemalloc.start()
big_list = [x * x for x in range(10_000_000)]
_,peak_list = tracemalloc.get_traced_memory()
tracemalloc.stop()
del big_list
tracemalloc.start()
big_gen = (x * x for x in range(10_000_000))
_,peak_gen = tracemalloc.get_traced_memory()
tracemalloc.stop()
print(f"列表推导式峰值内存: {peak_list / 1024 / 1024:.2f} MB")
print(f"生成器表达式峰值内存: {peak_gen / 1024 / 1024:.2f} MB")
```

```python
class FileHandler:
    def __init__(self,filename,mode='r'):
        self.filename = filename
        self.mode = mode
        self.file = None
    def __enter__(self):
        print(f"  [enter] 打开文件: {self.filename}")
        self.file = open(self.filename,self.mode)
        return self.file
    def __exit__(self,exc_type,exc_val,exc_tb):
        print(f"  [exit] 关闭文件: {self.filename}")
        print(f"  [exit] 异常信息: {exc_type}, {exc_val}")
        if self.file:
            self.file.close()
        return False
with FileHandler("/tmp/test.txt", 'w') as f:
    f.write("Hello, Context Manager!")
    print("  [body] 写入成功")
print()
try:
    with FileHandler("/tmp/test2.txt", 'w') as f:
        f.write("Before error")
        print("  [body] 即将抛出异常")
        raise ValueError("模拟异常")
except ValueError as e:
    print(f"  [caught] 捕获到异常: {e}")
```
