```python
def analyze_scores(scores):
    results = {
        'count' : len(scores),
        'average' : 0,
        'highest' : None,
        'lowest' : None,
        'above_90' : []
    }
    if not scores:
       return results
    total = 0
    highest_score = float('-inf')
    lowest_score = float('inf')
    for name,score in scores.items():
        total += score
        if score > highest_score:
            highest_score = score
        if score < lowest_score:
            lowest_score = score
        if score > 90:
            results['above_90'].append(name)
    results['average'] = round(total/len(scores),2)
    results['highest'] = highest_score
    results['lowest'] = lowest_score
    return results
```

```python
import random
def get_computer_choice():
    choices = ['石头','剪刀','布']
    return random.choice(choices)
def get_user_choice():
     return input("请输入'石头','剪刀','布'")
def decide_winner(user,computer):
    if user == computer:
        return "平局!"
    if ((user == "石头" and computer == "剪刀") or
        (user == "剪刀" and computer == "布") or 
        (user == "布" and computer == "石头")
    ):
        return "你赢了!"
    else:
        return "你输了!"
def play_game():
    user = get_user_choice()
    computer = get_computer_choice()
    print("你出了:",user)
    print("电脑出了:",computer)
    print(decide_winner(user,computer))
    print("游戏结束,谢谢参与!")
play_game()
```

```python
class Book:
    def __init__(self,title,author,pages):
        self.title = title
        self.author = author
        self.pages = pages
    def info(self):
        return f"{self.title} , {self.author} {self.pages}页"
    def __str__(self):
        return self.info()
    def is_long(self):
        return self.pages > 300
book1 = Book("三体","刘慈欣",400)
book2 = Book("小王子", "圣埃克苏佩里", 96)
print(book1)
print(book1.is_long())
print(book2)
print(book2.is_long())
```
