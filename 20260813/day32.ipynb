```python
import pandas as pd
import numpy as np
df = pd.DataFrame({
    "color": ["red", "blue", "green", "red", "blue", "green", "red", "yellow"],
    "size": ["S", "M", "L", "XL", "S", "M", "L", "XL"],
    "price": [10, 20, 30, 40, 15, 25, 35, 45],
})
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['size_encoded'] = le.fit_transform(df['size'])
print(dict(zip(le.classes_,le.transform(le.classes_))))

size_order = {"S": 1, "M": 2, "L": 3, "XL": 4}
df['size_ordered'] = df['size'].map(size_order)
print(df[['size', 'size_ordered']])

df_encoded = pd.get_dummies(df[['color','price']],columns=['color'],prefix='color')
print(df_encoded)

df['color_cat'] = df['color'].astype('category')
dummies = pd.get_dummies(df['color_cat'],prefix='color')
print(dummies)
```

```python
import pandas as pd
import numpy as np
np.random.seed(42)
df = pd.DataFrame({
    "salary": np.random.lognormal(mean=10.5, sigma=0.8, size=1000),
    "age": np.random.randint(18, 65, size=1000),
})
df['salary_zscore'] = (df['salary'] - df['salary'].mean()) / df['salary'].std()
print(df['salary_zscore'].mean(),df['salary_zscore'].std())
df['age_minmax'] = (df['age']-df['age'].min()) / (df['age'].max() - df['age'].min())
print(df['age_minmax'].min(),df['age_minmax'].max())
df['salary_log'] = np.log1p(df['salary'])
print(df['salary'].skew())
print(df['salary_log'].skew())
```
