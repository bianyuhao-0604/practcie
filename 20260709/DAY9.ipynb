```python
import requests
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry
import logging

logging.basicConfig(level=logging.INFO)

def create_robust_session() -> requests.Session:
    """创建一个带有重试机制和连接池的 Session"""
    session = requests.Session()
    
    # 配置重试策略：重试 3 次，退避因子 0.1，针对特定状态码重试
    retry_strategy = Retry(
        total=3,
        backoff_factor=0.1,  # 等待时间: 0.0s, 0.2s, 0.4s
        status_forcelist=[429, 500, 502, 503, 504],
        allowed_methods=["HEAD", "GET", "OPTIONS", "POST"]
    )
    
    # 挂载适配器到 http 和 https
    adapter = HTTPAdapter(max_retries=retry_strategy, pool_connections=10, pool_maxsize=10)
    session.mount("http://", adapter)
    session.mount("https://", adapter)
    
    # 设置通用 Headers
    session.headers.update({
        "User-Agent": "MyApp/1.0",
        "Accept": "application/json"
    })
    return session

def main():
    session = create_robust_session()
    base_url = "https://jsonplaceholder.typicode.com"
    
    try:
        # GET 请求（✅ 必须设置 timeout）
        logging.info("正在获取用户列表...")
        resp = session.get(f"{base_url}/users", timeout=(3.0, 10.0))  # (连接超时, 读取超时)
        resp.raise_for_status()  # 如果状态码不是 2xx，抛出 HTTPError
        users = resp.json()
        logging.info(f"获取到 {len(users)} 个用户")
        
        # POST 请求
        logging.info("正在创建新文章...")
        new_post = {"title": "Hello", "body": "World", "userId": 1}
        resp = session.post(f"{base_url}/posts", json=new_post, timeout=5.0)
        resp.raise_for_status()
        logging.info(f"创建成功，ID: {resp.json().get('id')}")
        
    except requests.exceptions.Timeout:
        logging.error("❌ 请求超时！请检查网络或增加 timeout 值")
    except requests.exceptions.HTTPError as e:
        logging.error(f"❌ HTTP 错误: {e.response.status_code} - {e.response.text}")
    except requests.exceptions.RequestException as e:
        logging.error(f"❌ 网络请求失败: {e}")

if __name__ == "__main__":
    main()
```
