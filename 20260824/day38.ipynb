```python
import time
import functools
def timer(func):
    @functools.wraps(func)
    def wrapper(*args,**kwargs):
        start = time.perf_counter()
        result = func(*args,**kwargs)
        elapsed = (time.perf_counter()-start) * 1000
        print(f"[{func.__name__}] executed in {elapsed:.3f}ms")
        return result
    return wrapper
@timer
def slow_sum(n):
    """计算 1 到 n 的和"""
    return sum(range(n))
print(slow_sum(1000000))
print(slow_sum.__name__)
print(slow_sum.__doc__)
```

```python
import functools
import time
def retry(max_attempts=3,delay=1,exceptions=(Exception,)):
    def decorator(func):
        @functools.wraps(func)
        def wrapper(*args,**kwargs):
            for attempt in range(1,max_attempts+1):
                try:
                    return func(*args,**kwargs)
                except exceptions as e:
                    if attempt == max_attempts:
                        raise
                    print(f"Attempt {attempt} failed: {e}. Retrying in {delay}s...")
                    time.sleep(delay)
        return wrapper
    return decorator
import random
@retry(max_attempts=3,delay=0.5,exceptions=(ValueError,))
def unstable_api():
    if random.random() < 0.7:
        raise ValueError("Network timeout")
    return "Success!"
print(unstable_api())
```
