```python
class Duck:
    def sound(self):
        return '嘎嘎嘎'
    def swim(self):
        return '鸭子在游泳'
class Dog:
    def sound(self):
        return '汪汪汪'
    def fetch(self):
        return '小狗在捡球'
class Robot:
    def sound(self):
        return '滋滋滋'
    def charge(self):
        return '机器人在充电'
class Person:
    def sound(self):
        return '你好'

def make_sound(thing):
    try:
        print(thing.sound())
    except AttributeError:
        print(f'{type(thing).__name__}不会发声')

animals_and_things = [Dog(),Duck(),Robot(),Person(),'字符串']
for item in animals_and_things:
    make_sound(item)
```

```python
class WeChatPay:
    def pay(self,amount):
        print(f'[微信支付]正在调起微信...')
        print(f'[微信支付]支付成功:￥{amount:.2f}')
        return True
class AliPay:
    def pay(self,amount):
        print(f'[支付宝支付]正在打开支付宝...')
        print(f'[支付宝]支付成功:￥{amount:.2f}')
        return True
class UnionPay:
    def pay(self,amount):
        print(f'[银联支付]正在刷卡...')
        print(f'[银联支付]支付成功:￥{amount:.2f}')
        return True
class ApplePay:
    def pay(self,amount):
        print(f'[ApplePay]正在打开ApplePay...')
        print(f'[ApplePay]支付成功:￥{amount:.2f}')
        return True

class ShoppingCart:
    def __init__(self):
        self.items = []
    def add_item(self,name,price):
        self.items.append({'name':name,'price':price})
    def get_total(self):
        return sum(item['price'] for item in self.items)
    def check_out(self,payment_method):
        total = self.get_total()

        print(f"\n=== 订单结算 ===")
        print(f"商品数量: {len(self.items)}")
        print(f"应付金额: ¥{total:.2f}")
        print("-" * 20)

        success = payment_method.pay(total)
        if success:
            print('支付成功')
            self.items.clear()
        else:
            print('支付失败')
cart = ShoppingCart()
cart.add_item('MacBook',15000)
cart.add_item('AirPods',2000)
cart.get_total()
cart.check_out(ApplePay())

```

```python
from abc import ABC,abstractmethod
class Worker(ABC):
    def __init__(self,name):
        self.name = name
    
    @abstractmethod
    def work(self):
        pass

    def rest(self):
        print(f'{self.name}正在休息')

class Programmer(Worker):
    def work(self):
        print(f'{self.name}正在疯狂敲代码')
class Designer(Worker):
    def work(self):
        print(f'{self.name}正在使用Figma画图')
class BadWorker(Worker):
    pass

p = Programmer('张三')
p.work()
p.rest()
d = Designer('李四')
d.work()
d.rest()
# b = BadWorker('王五')  无法被实例化，没有实现抽象方法
```

```python
from abc import ABC,abstractmethod
class Database(ABC):
    @property
    @abstractmethod
    def db_type(self):
        pass

    @abstractmethod
    def connect(self):
        pass
    @abstractmethod
    def disconnect(self):
        pass

    @classmethod
    @abstractmethod
    def get_default_port(cls):
        pass
class MySQLDatabase(Database):
    @property
    def db_type(self):
        return "MySQL"
    
    def connect(self):
        print(f'连接到{self.db_type}数据库(端口:{self.get_default_port()})...')
    
    def disconnect(self):
        print(f'断开{self.db_type}连接。')
    
    @classmethod
    def get_default_port(cls):
        return 3306

class RedisDatabase(Database):
    @property
    def db_type(self):
        return 'Redis'
    
    def connect(self):
        print(f'连接到{self.db_type}数据库(端口:{self.get_default_port()})...')
    def disconnect(self):
        print(f'断开{self.db_type}连接。')

    @classmethod
    def get_default_port(cls):
        return 6379

mysql = MySQLDatabase()
mysql.connect()
mysql.disconnect()
mysql.get_default_port()
r = RedisDatabase()
r.connect()
r.disconnect()
r.get_default_port()
```

```python
class CustomCollection:
    def __init__(self,items):
        self._items = items
    def __len__(self):
        return len(self._items)
def get_length(obj):
    try:
        return len(obj)
    except TypeError as e:
        raise TypeError(f"该对象不支持 len() 操作: {e}")
data = [1,2,3]
text = 'hello'
custom = CustomCollection(['a','b','c','d'])
print(get_length(data))
print(get_length(text))
print(get_length(custom))
```

```python
import logging
import datetime
logging.basicConfig(level=logging.INFO,format='%(message)s')
class LogMixin:
    def log_info(self,message):
        timestamp = datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')
        class_name = self.__class__.__name__
        logging.info(f'[{timestamp}][{class_name}INFO:{message}]')
class BaseEntiy:
    def save(self):
        pass
class User(BaseEntiy):
    def __init__(self,username):
        self.username = username
class Order(BaseEntiy):
    def __init__(self,order_id):
        self.order_id = order_id
class LoggableUser(LogMixin,User):
    def register(self):
        self.log_info(f'用户 {self.username} 注册成功！')
        self.save()
class LoggableOrder(LogMixin,Order):
    def create(self):
        self.log_info(f'订单 {self.order_id} 创建成功！')
        self.save()

u = LoggableUser('Alice')
u.register()
o = LoggableOrder('ORD-9527')
o.create()
```

```python
import json
class JsonMixin:
    def to_json(self):
        return json.dumps(self.to_dict(),ensure_ascii=False,indent=2)
class ValidationMixin:
    def validate(self):
        for field in getattr(self,'REQUIRED_FIELDS',[]):
            if not getattr(self,field,None):
                raise ValueError(f"缺少必填字段: {field}")
        print(f"[{self.__class__.__name__}] 数据校验通过 ✅")
class BaseModel:
    def to_dict(self):
        return {k:v for k,v in self.__dict__.items() if not k.startswith('_')}
class Product(JsonMixin,ValidationMixin,BaseModel):
    REQUIRED_FIELDS = ['name','price']
    def __init__(self,name,price,stock):
        self.name = name
        self.price = price
        self.stock = stock
# 测试
p = Product("机械键盘", 599, 100)  
p.validate()  
print(p.to_json())    
```

```python
class Countdown:
    def __init__(self,start):
        self.current = start
    def __iter__(self):
        return self
    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val
print("倒计时开始：")
for num in Countdown(18):
    print(num,end=' ')
print(list(Countdown(13)))
```

```python
class CaseInsenstiveDict:
    def __init__(self,data=None):
        self._store = {}
        if data:
            for k,v in data.items():
                self[k] = v
    def __setitem__(self,key,value):
        self._store[key.lower()] = (key,value)
    def __getitem__(self,key):
        return self._store[key.lower()][1]
    def __contains__(self,key):
        return key.lower() in self._store
    def __len__(self):
        return len(self._store)
    def __repr__(self):
        return str({k:v for k,v in self._store.values()})
headers = CaseInsenstiveDict({
    "Content-Type": "application/json",
    "Authorization": "Bearer token123"
})
print(headers['content-type'])
print(headers['CONTENT-TYPE'])

headers['accept'] = 'text/html'
print('ACCEPT' in headers)
print(len(headers))
print(headers)
```
