```python
import copy
original = [[1,2],[3,4],{"key":[5,6]}]
assigned = original
shallow = original.copy()
deep = copy.deepcopy(original)
original[0].append(99)
original[2]["key"].append(88)
print("original:",original)
print("assigned:",assigned)
print("shallow:",shallow)
print("deep:",deep)
```

```python
def add_item(item,lst=None):
    if lst is None:
        lst = []
    lst.append(item)
    return lst
print(add_item("apple"))
print(add_item("banana"))
print(add_item("cherry"))
```

```python
import copy
class ConfigManger:
    _defaults = {
        "db": {"host": "localhost", "port": 3306},
        "cache": {"ttl": 300, "max_size": 1000},
        "debug": False,
    }
    @classmethod
    def get_config(cls):
        return copy.deepcopy(cls._defaults)
config_a = ConfigManger.get_config()
config_b = ConfigManger.get_config()
config_a["db"]["port"] = 5432
config_a["debug"] = True
print("Config A:", config_a)
print("Config B:", config_b)
print("Defaults:", ConfigManger._defaults)
```

```python
import copy
class Node:
    def __init__(self,value,children=None):
        self.value = value
        self.children = children or []
    def __repr__(self):
        return f"Node({self.value}, children={self.children})"
root = Node("root", [Node("child1"), Node("child2")])
shallow_root = copy.copy(root)
shallow_root.children.append(Node("child3"))
deep_root = copy.deepcopy(root)
deep_root.children[0].value = "modified"
print("Original:",root)
print("Shallow:",  shallow_root)
print("Deep:",     deep_root)
```

```python
def my_range(start,stop=None,step=1):
    if stop is None:
        start,stop = 0,start
    if step == 0:
        raise ValueError("step must not be zero")
    current = start
    while (step > 0  and current < stop) or (step < 0 and current > stop):
        yield current
        current += step
print(list(my_range(5)))
print(list(my_range(2, 8)))
print(list(my_range(0, 10, 3)))
print(list(my_range(10, 0, -1)))
```

```python
def source():
    for i in range(1,20):
        yield i
def filter_even(gen):
    for x in gen:
        if x % 2 == 0:
            yield x
def square(gen):
    for x in gen:
        yield x * x
def take(n,gen):
    for i,x in enumerate(gen):
        if i >= n:
            break
        yield x
pipeline = take(5,square(filter_even(source())))
print(list(pipeline))
```

```python
def fibonacci():
    a,b = 0,1
    while True:
        yield a
        a,b = b,a+b
for num in fibonacci():
    if num > 10000:
        print(f"第一个超过10000的斐波那契数: {num}")
        break
```

```python
def match_pattern(pattern):
    print(f"开始过滤，关键词: {pattern}")
    while True:
        line = yield
        if pattern in line:
            print(f"  [匹配] {line.strip()}")
coro = match_pattern("ERROR")
next(coro)
logs = [
    "INFO: 服务启动成功",
    "ERROR: 数据库连接超时",
    "INFO: 用户登录",
    "ERROR: 文件未找到",
    "WARNING: 内存使用率 80%",
]
for log in logs:
    coro.send(log)
coro.close()
```
