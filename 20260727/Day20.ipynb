```python
from itertools import chain
list_a = [1,2,3]
list_b = [4,5,6]
list_c = [7,8,9]
for item in chain(list_a,list_b,list_c):
    print(item,end=' ')

mixed = chain('abc',[1,2,3],(True,False))
print(f'\n{list(mixed)}')

nested = [[1,2,3],[4,5,6],[7,8,9]]
flat = list(chain.from_iterable(nested))
print(flat)
```

```python
from itertools import islice,count
first_10 = list(islice(count(),10))
print(first_10)

result = list(islice(count(100,10),10))
print(result)

def fibonacci():
    a,b = 0,1
    while True:
        yield a
        a,b = b,a+b

fib_slice = list(islice(fibonacci(),5,10))
print(fib_slice)
```

```python
from itertools import groupby
from operator import itemgetter
students = [
    {"name": "小明", "grade": "A"},
    {"name": "小红", "grade": "B"},
    {"name": "小刚", "grade": "A"},
    {"name": "小美", "grade": "C"},
    {"name": "小李", "grade": "B"},
    {"name": "小王", "grade": "A"},
]
students_sorted = sorted(students,key=itemgetter('grade'))
for grade,group in groupby(students_sorted,key=itemgetter('grade')):
    names = [s['name'] for s in group]
    print(f'{grade}:{names}')

text = 'aaabbcccaaa'
print(text)
for char,group in groupby(text):
    count = len(list(group))
    print(f'{char}×{count}')
```

```python
from collections import Counter
text = 'the cat sat on the mat the cat'
words = text.split()
word_count = Counter(words)
print(word_count)


print(word_count.most_common(2))

print(word_count['dog'])

a = Counter('aabbbcccc')
b = Counter('abbc')
print(a)
print(b)
print(a+b)
print(a-b)
print(a&b)
print(a|b)

nums = [1,2,2,3,3,3,4]
duplicates = [item for item,count in Counter(nums).items() if count > 1]
print(duplicates)
```

```python
from collections import defaultdict
groups_old = {}
pairs = [("水果", "苹果"), ("蔬菜", "白菜"), ("水果", "香蕉"), ("蔬菜", "萝卜")]
for category,item in pairs:
    if category not in groups_old:
        groups_old[category] = []
    groups_old[category].append(item)
groups = defaultdict(list)
for category,item in pairs:
    groups[category].append(item)
print(dict(groups))    
    
counter = defaultdict(int)
counter['apple'] += 1
counter['apple'] += 1
counter['banana'] += 1
print(dict(counter))

nested = defaultdict(lambda:defaultdict(int))
nested['2024']['January'] = 31
nested['2024']['February'] = 29
print(nested['2024']['March'])

words = ['apple','avocado','banana','blueberry','cherry']
by_letter = defaultdict(list)
for w in words:
    by_letter[w[0]].append(w)
for letter in sorted(by_letter):
    print(f'{letter}:{by_letter[letter]}')
```
