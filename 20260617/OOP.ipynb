```python
class Dog:
    def __init__(self,name):
        self.name = name
    def bark(self):
        return f"{self.name}:汪汪"
dog1 = Dog("小黄")
print(dog1.bark())
```

```python
class Rectangle:
    def __init__(self,width,height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
    def perimeter(self):
        return 2 * (self.width + self.height)
rect1 = Rectangle(4,5)
print(f"面积:{rect1.area()}, 周长: {rect1.perimeter()}")
```

```python
class Student:
    def __init__(self,name,lesson1_score,lesson2_score,lesson3_score):
        self.name = name
        self.lesson1_score = lesson1_score
        self.lesson2_score = lesson2_score
        self.lesson3_score = lesson3_score
    def average_score(self):
        return (self.lesson1_score + self.lesson2_score + self.lesson3_score)/3
student1 = Student("张飞",90,85,95)
print(student1.average_score())
```

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance = balance
    def deposit(self,amount):
        self.balance += amount
    def withdraw(self,amount):
        self.balance -= amount
```

```python
class Employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    @property
    def salary(self):
        return self._salary
    @salary.setter
    def salary(self,value):
        if value < 0:
            raise ValueError("Salary cannot be negative")
        self._salary = value
```

```python
class Animal:
    def __init__(self,name):
        self.name = name
    def speak(self):
        raise NotImplementedError("Subclasses must implement this method")
class Dog(Animal):
    def speak(self):
        return f"{self.name}汪汪"
class Cat(Animal):
    def speak(self):
        return f"{self.name}喵喵"
dog1 = Dog("小黄")
cat1 = Cat("小花")
print(dog1.speak())
print(cat1.speak())
```

```python
from abc import ABC,abstractmethod
import math
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2
class Square(Shape):
    def __init__(self,side):
        self.side = side
    def area(self):
        return self.side ** 2


```

```python
class Engine:
    def start(self):
        print("发动机启动")
    def stop(self):
        print("发动机停止")
class Car:
    def __init__(self,engine):
        self.engine = engine
    def drive(self):
        self.engine.start()
engine = Engine()
car = Car(engine)
print(car.drive())
    
```
