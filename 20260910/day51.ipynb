```python
import math
def distance(p1,p2):
    x1,y1 = p1
    x2,y2 = p2
    return math.sqrt((x2-x1)**2 + (y2-y1)**2)
point_a = (0,0)
point_b = (3,4)
d = distance(point_a,point_b)
print(f"\n{point_a}到{point_b}的距离{d}")
point_c = (1,1)
point_d = (4,5)
d2 = distance(point_c,point_d)
print(f"{point_c}到{point_d}的距离{d2}")
```

```python
def analyze_numbers(numbers):
    return (
        min(numbers),
        max(numbers),
        sum(numbers),
        sum(numbers) / len(numbers),
        len(numbers)
    )
scores = [78, 92, 65, 88, 45, 91, 73, 56, 82, 97]
result = analyze_numbers(scores)
print(f"\n统计结果(元组):{result}")
minimum,maximum,total,average,count = analyze_numbers(scores)
print(f"最小值:{minimum}")
print(f"最大值:{maximum}")
print(f"总和:{total}")
print(f"平均值:{average:.1f}")
print(f"个数:{count}")
```
