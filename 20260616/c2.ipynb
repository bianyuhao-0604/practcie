```python
from pathlib import Path
import cv2
input_dir = Path('images')
output_dir = Path('results')
output_dir.mkdir(exist_ok=True)
for path in input_dir.glob('*.jpg'):
    img = cv2.imread(str(path))
    if img is None:
       print('读取失败:', path)
       continue
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    out_path = output_dir / f'{path.stem}_gray.png'
    cv2.imwrite(str(out_path), gray)
```

```python
import os
import random
import shutil
from pathlib import Path

# ---------- 可配置参数 ----------
SOURCE_DIR = './images'
TARGET_ROOT = 'dogs'
NUM_BREEDS = 10                 # 要提取的犬种数量（实际可能少于白名单数量）
IMAGES_PER_BREED = 10
IMAGE_EXTENSIONS = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff'}

# ========== 关键：定义你需要的犬种名称（与文件名前缀一致） ==========
BREED_WHITELIST = {
    'beagle', 'english', 'boxer', 'german', 'great', 
    'havanese', 'yorkshire', 'wheaten', 'keeshond', 'chihuahua',
    # 在这里添加你所有可能出现的犬种前缀
}
# 如果你不想手动列出全部，也可以用黑名单排除 'cat', 'bird' 等
# ===================================================================
# --------------------------------

def split_breed(filename):
    stem = Path(filename).stem
    parts = stem.split('_')
    return parts[0] if len(parts) > 1 else stem

def main():
    all_files = []
    for f in os.listdir(SOURCE_DIR):
        path = Path(f)
        if path.suffix.lower() in IMAGE_EXTENSIONS:
            all_files.append(f)

    print(f"共发现 {len(all_files)} 张图片")

    # 按品种分组
    breed_dict = {}
    for f in all_files:
        breed = split_breed(f)
        # 只保留白名单中的品种（如果使用了白名单）
        if breed in BREED_WHITELIST:   
            breed_dict.setdefault(breed, []).append(f)

    print(f"白名单内品种有 {len(breed_dict)} 个")

    # 筛选出图片数量 >= IMAGES_PER_BREED 的品种
    candidate_breeds = [b for b, files in breed_dict.items() if len(files) >= IMAGES_PER_BREED]
    if len(candidate_breeds) == 0:
        print("错误：没有满足条件的犬种，请检查白名单或图片命名")
        return

    # 随机选择 NUM_BREEDS 个（如果不足则全选）
    if len(candidate_breeds) >= NUM_BREEDS:
        selected_breeds = random.sample(candidate_breeds, NUM_BREEDS)
    else:
        selected_breeds = candidate_breeds
        print(f"警告：满足条件的犬种仅有 {len(selected_breeds)} 个，将全部提取")

    print(f"选中品种：{selected_breeds}")

    os.makedirs(TARGET_ROOT, exist_ok=True)

    for breed in selected_breeds:
        breed_files = breed_dict[breed]
        chosen = random.sample(breed_files, IMAGES_PER_BREED)
        breed_dir = os.path.join(TARGET_ROOT, breed)
        os.makedirs(breed_dir, exist_ok=True)
        for filename in chosen:
            shutil.copy2(os.path.join(SOURCE_DIR, filename), os.path.join(breed_dir, filename))
        print(f"品种 '{breed}' 已提取 {len(chosen)} 张图片")

    print("全部完成！")

if __name__ == '__main__':
    main()
```
