```python
def get_squares(n):
    for i in range(n):
        yield i ** 2
print(list(get_squares(5)))

for sq in get_squares(5):
    print(sq)
```

```python
def fibonacci():
    a,b = 0,1
    while True:
        yield a
        a,b = b,a+b
fib = fibonacci()
first_10 = [next(fib) for _ in range(10)]
print(first_10)

for num in fib:
    if num > 1000000000000:
        print(num)
        break
```

```python
def read_lines(filename):
    with open(filename,'r',encoding='utf-8') as f:
        for line in f:
            yield line.strip()
def search_file(filename,keyword):
    for line_num,line in enumerate(read_lines(filename),1):
        if keyword in line:
            yield line_num,line
with open('text.txt','w',encoding='utf-8') as f:
    for i in range(1000):
        status = 'ERROR' if i % 100 == 0 else 'OK'
        f.write(f"第{i}行: {status} - 一切正常\n")

for num,line in search_file('text.txt','ERROR'):
    print(f'行{num}:{line}')
import os
os.remove('text.txt')
```

```python
students = [
    {"name": "小明", "age": 18, "score": 85},
    {"name": "小红", "age": 17, "score": 92},
    {"name": "小刚", "age": 19, "score": 60},
    {"name": "小美", "age": 18, "score": 78},
    {"name": "小李", "age": 20, "score": 95},
]
def adults(data):
    for s in data:
        if s['age'] >= 18:
            yield s
def high_scores(data,threshold=80):
    for s in data:
        if s['score'] >= threshold:
            yield s
results = high_scores(adults(students),80)
for s in results:
    print(f"{s['name']}:{s['age']}")
```

```python
def paginate(data,page_size):
    for i in range(0,len(data),page_size):
        yield data[i:i+page_size]
all_items = list(range(1,26))
for page_num,page in enumerate(paginate(all_items,7),1):
    print(f"  第{page_num}页: {page}")

pages = paginate(all_items,10)
next(pages)
page2 = next(pages)
print(f"\n第2页: {page2}")
```

```python
def generate_users(n):
    for i in range(1,n+1):
        yield {"id": i, "name": f"用户{i}", "score": i % 100}
def batch_process(data,batch_size=1000):
    batch = []
    for item in data:
        batch.append(item)
        if len(batch) >= batch_size:
            yield batch
            batch =[]
        if batch:
            yield batch
total_processed = 0
for batch_num,batch in enumerate(batch_process(generate_users(1000000)),1):
    total_processed += len(batch)
    if batch_num % 100 == 0:
        print(f"  已处理 {total_processed:>10,} 条 ({total_processed/10000:.0f}%)")
print(f"\n✅ 完成！共处理 {total_processed:,} 条数据")
print("💡 全程内存中最多只有 1000 条数据（一个 batch 的大小）")
```
