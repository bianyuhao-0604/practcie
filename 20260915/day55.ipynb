```python
class_a = {"小明", "小红", "小刚", "小丽"}
class_b = {"小刚", "小丽", "小华", "小陈"}
both = class_a & class_b
print(both)
both2 = class_a.intersection(class_b)
print("intersection:",both2)
all_students = class_a | class_b
print("全部的学生",all_students)
all_students2 = class_a.union(class_b)
print("union",all_students2)
only_a = class_a - class_b
print("只在A班的学生",only_a)
only_a2 = class_a.difference(class_b)
print("difference",only_a2)
sym_diff = class_a ^ class_b
print("只在一个班的学生(对称差集):",sym_diff)
sym_diff2 = class_a.symmetric_difference(class_b)
print("symmetric_difference:",sym_diff2)
```

```python
friends = {
    "Alice": {"Bob", "Charlie", "David"},
    "Bob": {"Alice", "Charlie", "Eve"},
}
class FriendRecommendation:
    def __init__(self):
        self.friends = {}
    def add_friendship(self,user1,user2):
        self.friends.setdefault(user1,set()).add(user2)
        self.friends.setdefault(user2,set()).add(user1)
    def get_friends(self,user):
        return self.friends.get(user,set())
    def recommend_friends(self,user):
        if user not in self.friends:
            return set()
        direct_friends = self.friends[user]
        friends_of_friends = set()
        for friend in direct_friends:
           friends_of_friends.update(self.friends.get(friend,set()))
        recommendations = friends_of_friends - direct_friends - {user}
        return recommendations
    def common_friends_count(self,user1,user2):
        friends1 = self.friends.get(user1,set())
        friends2 = self.friends.get(user2,set())
        return len(friends1 & friends2)
fr = FriendRecommendation()

fr.add_friendship("Alice", "Bob")
fr.add_friendship("Alice", "Charlie")
fr.add_friendship("Bob", "Charlie")
fr.add_friendship("Bob", "David")
fr.add_friendship("Charlie", "Eve")
fr.add_friendship("David", "Eve")
print("Alice的好友:",fr.get_friends("Alice"))
print("推荐给Alice的人:",fr.recommend_friends("Alice"))
print("Alice和Bob的共同好友数:",fr.common_friends_count("Alice","Bob"))
```
