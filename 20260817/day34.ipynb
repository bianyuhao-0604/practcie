```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import cross_val_score,StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
import numpy as np
data = load_breast_cancer()
X,y = data.data,data.target
rf = RandomForestClassifier(n_estimators=100,random_state=42)
score = cross_val_score(rf,X,y,cv=5,scoring='accuracy')
print(f"随机森林 5 折交叉验证:")
print(f"  每折准确率: {scores.round(4)}")
print(f"  平均: {scores.mean():.4f} ± {scores.std():.4f}")
pipeline_lr = Pipeline([
    ('scaler',StandardScaler()),
    ('clf',LogisticRegression(max_iter=10000,random_state=42))
])
pipeline_rf = Pipeline([
    ('scaler',StandardScaler()),
    ('rf',RandomForestClassifier(n_estimators=100,random_state=42))
])
cv = StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
for name,pipe in [('逻辑回归',pipeline_lr),('随机森林',pipeline_rf)]:
    score = cross_val_score(pipe,X,y,cv=cv,scoring='accuracy')
    print(f"\n{name} (Pipeline + 5折CV):")
    print(f"  每折: {scores.round(4)}")
    print(f"  平均: {scores.mean():.4f} ± {scores.std():.4f}")
```

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np
X,y_true = make_blobs(n_samples=300,centers=3,random_state=42,cluster_std=1.0)
kmeans = KMeans(n_clusters=3,random_state=42,n_init=10)
y_pred = kmeans.fit_predict(X)
print(f"聚类中心:\n{kmeans.cluster_centers_.round(2)}")
print(f"轮廓系数: {silhouette_score(X, y_pred):.4f}")
inertias = []
K_range = range(2,11)
for k in K_range:
    km = KMeans(n_clusters=k,random_state=42,n_init=10)
    km.fit(X)
    inertias.append(km.inertia_)
    print(f"K={k}: Inertia={km.inertia_:.1f}, 轮廓系数={silhouette_score(X, km.labels_):.4f}")
```
