```python
import torch

# 3 张图: 3, 4, 5 节点. N_max = 5
B, N_max = 3, 5
sizes = [3, 4, 5]

mask = torch.zeros(B, N_max, dtype=torch.bool)
for i, n in enumerate(sizes):
    mask[i, :n] = True   # 前 n 个位置设为 True

print('Mask:')
print(mask.int())  # int() 让 0/1 更直观
print(f'每张图的真实节点数: {mask.sum(dim=1).tolist()}')
```

```python
import torch
A = torch.zeros(5,5)
edges = [(0,1),(0,2),(1,2),(2,3),(3,4)]
for u,v in edges:
    A[u,v] = 1
    A[v,u] = 1
print(A)
print(A.sum().item())
```

```python
import torch
edges = [(0,1),(0,2),(1,2),(2,3),(3,4)]
directed_edges = edges + [(v,u) for u,v in edges]
edge_index = torch.tensor(directed_edges,dtype=torch.long).t()
print(edge_index.shape)
print(edge_index)
```

```python
import torch
from torch_geometric.data import Data
X = torch.randn(5,3)
edge_index = torch.tensor([[0,1,0,2,1,2,2,3,3,4],
                          [1,0,2,0,2,1,3,2,4,3]],dtype=torch.long)
g = Data(x=X,edge_index=edge_index)
print(g.num_nodes)
print(g.num_edges)
print(g.has_self_loops())
```
