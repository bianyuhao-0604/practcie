```python
school = {
    "A班": {
        "班主任": "张老师",
        "学生": {
            "小明": {"语文": 85, "数学": 92},
            "小红": {"语文": 90, "数学": 88},
        }
    },
    "B班": {
        "班主任": "李老师",
        "学生": {
            "小刚": {"语文": 72, "数学": 65},
            "小丽": {"语文": 95, "数学": 97},
        }
    }
}
print(school['A班']['班主任'])
print(school['A班']['学生']['小明'])
print(school['B班']['学生']['小丽']['语文'])
school['A班']['学生']['小明']['英语'] = 78
school['B班']['班主任'] = '王老师'
print(school['A班']['学生'])
print(school['B班']['班主任'])
for class_name,class_info in school.items():
    print(f"\n==={class_name}(班主任:{class_info['班主任']})===")
    for student_name,student_scores in class_info['学生'].items():
        total = sum(student_scores.values())
        avg = total / len(student_scores)
        subjects = ", ".join(f"{k}={v}" for k,v in student_scores.items())
        print(f"  {student_name}:{subjects},总分={total},平均={avg:.1f}")
```

```python
data = {
    "user": {
        "name": "小明",
        "address": {
            "city": "苏州"
        }
    }
}
street = data.get("user",{}).get("address",{}).get("street","未知")
print(f"街道:{street}")
def deep_get(d,keys,default=None):
    for key in keys:
        if isinstance(d,dict):
            d = d.get(key,default)
        else:
            return default
    return d
print(deep_get(data,["user","address","city"]))
print(deep_get(data,["user","address","street"]))
print(deep_get(data,["user","address","street"],"未知"))
print(deep_get(data,{"user","address","street"},"无"))
```

```python
dict1 = {"a": 1,"b":2,"c":3}
dict2 = {"c":30,"d":4,"e":5}
merged1 = dict1.copy()
merged1.update(dict2)
print("update():",merged1)
merged2 = dict1 | dict2
print("|运算符:",merged2)
merged3 = {**dict1,**dict2}
print("**解包:",merged3)
def merge_dicts(d1,d2,strategy="override"):
    result = d1.copy()
    for key,value in d2.items():
        if key in result:
            if strategy == "override":
                result[key] = value
            elif strategy == "keep":
                pass
            elif strategy == "sum":
                result[key] = result[key] + value
            elif strategy == "list":
                result[key] = [result[key],value]
        else:
            result[key]  = value
    return result
d1 = {"a":1,"b":2}
d2 = {"b":3,"c":4}
print("覆盖策略:",merge_dicts(d1,d2,"override"))       
print("保留策略:",merge_dicts(d1,d2,"keep"))
print("求和策略:",merge_dicts(d1,d2,"sum"))
```

```python
from collections import defaultdict
word_count = defaultdict(int)
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
for word in words:
    word_count[word] += 1
print("词频:",dict(word_count))
students = [
    ("A班", "小明"),
    ("B班", "小红"),
    ("A班", "小刚"),
    ("C班", "小丽"),
    ("B班", "小华"),
]
class_groups = defaultdict(list)
for cls,name in students:
    class_groups[cls].append(name)
print("分组:",dict(class_groups))
tags = [
    ("Python", "编程"),
    ("Java", "编程"),
    ("Python", "脚本"),
    ("Java", "企业"),
    ("Python", "数据"),
]
tag_groups = defaultdict(set)
for lang,tag in tags:
    tag_groups[lang].add(tag)
print("标签:",dict(tag_groups))
def default_student():
    return {"语文":0,"数学":0,"英语":0}
grade_book = defaultdict(default_student)
grade_book["小明"]["语文"] = 85
grade_book["小明"]["数学"] = 92
print("成绩:",dict(grade_book))
```

```python
from collections import Counter
words = ["apple","banana","apple","cherry","banana","apple"]
counter = Counter(words)
print("计数:",counter)
print("类型:",type(counter))
text = "abracadabra"
char_count = Counter(text)
print("字符计数:",char_count)
print("最常见的2个:",counter.most_common(2))
print("apple的次数:",counter["apple"])
print("不存在的不报错:",counter["grape"])
c1 = Counter("abracadabra")
c2 = Counter("alacazam")
print("合并(相加):",c1+c2)
print("差集(相减):",c1-c2)
print("交集(取小):",c1&c2)
print("并集(取大):",c1 | c2)
article = """
Python is great. Python is easy to learn.
Many people love Python. Python is powerful.
"""
words = article.lower().replace(".","").split()
top3 = Counter(words).most_common(3)
print("\n最常见的几个词:",top3)
```

```python
scores = {"小明": 85, "小红": 92, "小刚": 58, "小丽": 95, "小华": 45}
passed = {name:scores for name,score in scores.items() if score >= 60}
print("及格:",passed)
failed = {name:score for name,score in scores.items() if score < 60}
print("不及格:",failed)
bonus = {name:score + 5 for name,score in scores.items()}
print("加5分:",bonus)
def to_grade(score):
    if score >= 90: return "A"
    if score >= 80: return "B"
    if score >= 70: return "C"
    if score >= 60: return "D"
    return "F"
grades = {name:to_grade(score) for name,score in scores.items()}
print("等级:",grades)
```
