```python
sentence = "Hello, World! This is a Python for loop practice."
vowel_count = 0
vowels = "aeiou"
list = []
for char in sentence:
    if char in vowels:
        vowel_count += 1
        list.append(char)
print(f"句子中的元音字母为：{list},个数为：{vowel_count}")
        
```
