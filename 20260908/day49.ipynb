```python
import random


def get_valid_number(prompt, min_val, max_val):
    """获取用户输入的数字，确保在合法范围内"""
    while True:
        text = input(prompt)
        try:
            num = int(text)
        except ValueError:
            print("⚠ 输入无效，请输入一个整数！")
            continue
        if num < min_val or num > max_val:
            print(f"⚠ 请输入 {min_val} 到 {max_val} 之间的数字！")
            continue
        return num


def play_one_game(difficulty):
    """进行一局猜数字游戏，返回猜测次数"""
    min_val = difficulty["min"]
    max_val = difficulty["max"]
    target = random.randint(min_val, max_val)
    attempts = 0

    print(f"\n🎮 游戏开始！我已经想好了一个 {min_val}-{max_val} 之间的数字")
    print("-" * 45)

    while True:
        guess = get_valid_number(f"第 {attempts + 1} 次猜测：", min_val, max_val)
        attempts += 1

        if guess < target:
            print("  📈 太小了，再大一点！")
        elif guess > target:
            print("  📉 太大了，再小一点！")
        else:
            print(f"\n  🎉 恭喜你！答案就是 {target}，你用了 {attempts} 次猜对！")
            break

    if attempts == 1:
        print("  ⭐ 评价：一发入魂！你是天选之人！")
    elif attempts <= 5:
        print("  ⭐ 评价：天才！太厉害了！")
    elif attempts <= 10:
        print("  ⭐ 评价：不错，继续加油！")
    elif attempts <= 15:
        print("  ⭐ 评价：还行，还有进步空间")
    else:
        print("  ⭐ 评价：下次一定行！多练练吧")

    return attempts


def choose_difficulty():
    """让玩家选择难度"""
    difficulties = {
        "1": {"name": "简单（1-50）",   "min": 1, "max": 50},
        "2": {"name": "普通（1-100）",  "min": 1, "max": 100},
        "3": {"name": "困难（1-500）",  "min": 1, "max": 500},
    }

    print("\n请选择难度：")
    print("  1. 简单（1-50）")
    print("  2. 普通（1-100）")
    print("  3. 困难（1-500）")

    while True:
        choice = input("请输入 1/2/3：").strip()
        if choice in difficulties:
            selected = difficulties[choice]
            print(f"已选择：{selected['name']}")
            return selected
        print("⚠ 请输入 1、2 或 3")


def show_history(scores):
    """显示历史成绩统计"""
    if not scores:
        print("\n还没有游戏记录")
        return

    print("\n" + "=" * 45)
    print("           📊 历史成绩统计")
    print("=" * 45)

    for i, score in enumerate(scores, start=1):
        print(f"  第{i}局：{score['difficulty']}难度，猜测 {score['attempts']} 次")

    all_attempts = [s["attempts"] for s in scores]
    total_games = len(scores)
    avg = sum(all_attempts) / total_games
    best = min(all_attempts)
    worst = max(all_attempts)

    print("-" * 45)
    print(f"  总局数：{total_games}")
    print(f"  平均猜测次数：{avg:.1f}")
    print(f"  最佳成绩：{best} 次")
    print(f"  最差成绩：{worst} 次")
    print("=" * 45)


def main():
    """猜数字游戏主程序"""
    scores = []

    print("=" * 45)
    print("       🎯 猜数字游戏 v2.0")
    print("=" * 45)

    while True:
        print("\n--- 主菜单 ---")
        print("  1. 开始新游戏")
        print("  2. 查看历史记录")
        print("  3. 退出游戏")

        menu = input("请选择（1/2/3）：").strip()

        if menu == "1":
            difficulty = choose_difficulty()
            attempts = play_one_game(difficulty)
            scores.append({
                "difficulty": difficulty["name"],
                "attempts": attempts
            })
        elif menu == "2":
            show_history(scores)
        elif menu == "3":
            if scores:
                show_history(scores)
            print("\n感谢游玩，再见！👋")
            break
        else:
            print("⚠ 请输入 1、2 或 3")


main()
```
