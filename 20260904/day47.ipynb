```python
def check_password(pwd):
    length_ok = len(pwd) >= 8
    has_upper = pwd != pwd.lower()
    has_lower = pwd != pwd.upper()
    has_digit = any(c.isdigit for c in pwd)
    score = sum([length_ok,has_upper,has_lower,has_digit])
    if score == 4:
        level = '强'
    elif score == 4:
        level = '中'
    else:
        level = '弱'
    print(level)
check_password("abc")
check_password("hello123")
check_password("Hello123")
check_password("Hello, World! 2026")
```

```python
def check(word) -> bool:
    cleanned = word.lower().replace(" ","")
    return cleanned == cleanned[::-1]
print(check("asdasdsad"))
```

```python
def func(text):
    total_chars_no_spaces = len(text.replace(" ",""))
    total_words = len(text.split())
    import re
    from collections import Counter
    cleanned = re.sub(r'[^a-zA-Z\s]','',text.lower())
    word_list = cleanned.split()
    if word_list:
        word_counts = Counter(word_list)
        most_common_word,most_common_count = word_counts.most_common(1)[0]
    else:
        most_common_word,most_common_count = None,0
    print(total_chars_no_spaces)
    print(total_words)
    print(most_common_word,most_common_count)
sample_text = "Hello, world! Hello everyone. This is a test, test world."
func(sample_text)
```

```python
def encrypt(text):
    result = []
    for ch in text:
        if 'A' <= ch <= 'Z':
            shifted = (ord(ch)-ord('A')+3) % 26 + ord('A')
            result.append(chr(shifted))
        elif 'a' <= ch <= 'z':
            shifted = (ord(ch)-ord('a')+3) % 26 + ord('a')
            result.append(chr(shifted))
        else:
            result.append(ch)
    return ''.join(result)
def decrypt(text):
    result = []
    for ch in text:
        if 'A' <= ch <= 'Z':
            shifted = (ord(ch)-ord('A')-3) % 26 + ord('A')
            result.append(chr(shifted))
        elif 'a' <= ch <= 'z':
            shifted = (ord(ch)-ord('a')-3) % 26 + ord('a')
            result.append(chr(shifted))
        else:
            result.append(ch)
    return ''.join(result)
plain = "Hello, World! XYZ xyz"
cipher = encrypt(plain)
decrypted = decrypt(cipher)

print("原文:", plain)
print("加密:", cipher)
print("解密:", decrypted)
```
