```python
def make_student(name,age,score):
    return {"name":name,"age":age,"score":score}
def get_grade(s):
    if s["score"] >= 90:return "A"
    elif s["score"] >= 80:return "B"
    else: return "C"
def birthday(s):
    s["age"] += 1
    return s
stu = make_student("小明",18,92)
print(get_grade(stu))
```

```python
class Student:
    def __init__(self,name,age,score):
        self.name = name
        self.age = age
        self.score = score
    def get_grade(self):
        if self.score >= 90:return "A"
        elif self.score >= 80:return "B"
        else:return "C"
    def birthday(self):
        self.age += 1
stu = Student("小明",18,92)
print(stu.get_grade())
stu.birthday()
print(stu.age)
```

```python
class Dog:
    def __init__(self,name,breed="中华田园犬"):
        self.name = name
        self.breed = breed
        self.energy = 100
    def bark(self):
        print(f"{self.name}：汪汪汪！")
    def run(self,distance):
        self.energy -= distance * 2
        print(f"{self.name}跑了{distance}米，剩余体力{self.energy}")
    def eat(self,food):
        self.energy += 30
        print(f"{self.name}吃了{food}，体力恢复到{self.energy}")
dog1 = Dog("旺财")
dog2 = Dog("小白","萨摩耶")
print(dog1.name)
print(dog2.breed)
dog1.bark()
dog2.run(10)
dog1.eat("骨头")
print(dog1.energy)
print(dog2.energy)
dog1.owner = "小明"
print(dog1.owner)
del dog1.owner
```

```python
import random
from datetime import datetime
class Player:
    game_name = "勇者斗恶龙"
    def __init__(self,name,hp=100):
        self.name = name
        self.hp = hp
        self.level = 1
    def attack(self,target):
        damage = random.randint(10,20) * self.level
        target.hp -= damage
        print(f"{self.name}攻击{target.name},造成{damage}点伤害")
    @classmethod
    def from_string(cls,data_str):
        name,hp,level = data_str.split(",")
        p = cls(name,int(hp))
        p.level = int(level)
        return p 
    @classmethod
    def create_healer(cls,name):
        p = cls(name,hp=150)
        p.role = "治疗者"
        return p
    @staticmethod
    def is_valid_name(name):
        return isinstance(name,str) and 1 <= len(name) <= 10
    def __str__(self):
        return f"{self.name}(HP:{self.hp},LV:{self.level})"
p1 = Player("战士小明")
p2 = Player.from_string("法师小红,80,3")
p3 = Player.create_healer("牧师小刚")
print(p1,p2,p3,sep="\n")
p1.attack(p2)
print(p2)
print(Player.is_valid_name("小明"))
print(p1.is_valid_name("小明"))
print(Player.game_name)
```

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.name = owner
        self._balance = balance
        self.__secret = "密码123"
    def get_balance(self):
        return self._balance
    def deposit(self,amount):
        if amount <= 0:
            raise ValueError("存款金额必须大于0")
        self._balance += amount
        print(f"存入{amount}，余额{self._balance}")
    def withdraw(self,amount):
        if amount <= 0:
            raise ValueError("取款金额必须大于0")
        if amount > self._balance:
            raise ValueError("余额不足")
        self._balance -= amount
        print(f"取出{amount}，余额{self._balance}")
acc = BankAccount("小明", 1000)
acc.deposit(500)
acc.withdraw(200)
print(acc.get_balance())
```

```python
class Temperature:
    def __init__(self,celsius=0):
        self._celsius = celsius
    @property
    def celsius(self):
        return self._celsius
    @celsius.setter
    def celsius(self,value):
        if value < -273.15:
            raise ValueError("温度不能低于绝对零度")
        self._celsius = value
    @property
    def fahrenheit(self):
        return self._celsius * 9 / 5 + 32
    @fahrenheit.setter
    def fahrenheit(self,value):
        self._celsius = (value-32) * 5 / 9
    def __str__(self):
        return f"{self._celsius}°C / {self.fahrenheit:.1f}°F"
t = Temperature(25)
print(t.celsius)
print(t.fahrenheit)
t.celsius = 100
print(t)
t.fahrenheit = 32
print(t.celsius)
```

```python
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
    def __str__(self):
        return f"({self.x}, {self.y})"
    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)
    def __sub__(self,other):
        return Vector(self.x-other.x,self.y-other.y)
    def __mul__(self,scalar):
        return Vector(self.x * scalar,self.y * scalar)
    __rmul__ = __mul__
    def __truediv__(self,scalar):
        return Vector(self.x / scalar,self.y / scalar)
    def __neg__(self):
        return Vector(-self.x,-self.y)
    def __eq__(self,other):
        if not isinstance(other,Vector):
            return NotImplemented
        return self.x == other.x and self.y == other.y
    def __lt__(self,other):
        return (self.x**2+self.y**2) < (other.x**2+other.y**2)
    def __abs__(self):
        return (self.x**2+self.y**2) ** 0.5
    def __len__(self):
        return 2
    def __bool__(self):
        return self.x != 0 or self.y != 0 
    def __iter__(self):
        yield self.x
        yield self.y
    def __getitem__(self,index):
        return [self.x,self.y][index]
v1 = Vector(3,4)
v2 = Vector(1,2)
print(v1)
print(repr(v1))
print(v1+v2)
print(v1*3)
print(3*v1)
print(-v1)
print(abs(v1))
print(v1==v2)
print(v1<v2)
print(bool(Vector(0,0)))
x,y = v1
print(x,y)
print(v1[0],v1[1])
vecs = [Vector(3,4),Vector(1,1),Vector(2,3)]
print(sorted(vecs))
```

```python
class Animal:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def speak(self):
        raise NotImplementedError("子类必须实现 speak 方法")
    def info(self):
        return f"{self.name},{self.age}岁"
    def __repr__(self):
        return f"{self.__class__.__name__}({self.name})"
class Dog(Animal):
    species = "犬科"
    def __init__(self,name,age,breed):
        super().__init__(name,age)
        self.breed = breed
    def speak(self):
        return f"{self.name}:汪汪!"
    def fetch(self):
        return f"{self.name}捡回了球"
    def info(self):
        base = super().info()
        return f"{base},品种:{self.breed}"
class Cat(Animal):
    def __init__(self,name,age,indoor=True):
        super().__init__(name,age)
        self.indoor = indoor
    def speak(self):
        return f"{self.name}:喵"
    def info(self):
        place = "室内猫" if self.indoor else "室外猫"
        return f"{super().info()},{place}"
d = Dog("旺财",3,"柴犬")
c = Cat("咪咪",2,indoor=True)
print(d.speak())
print(c.speak())
print(d.info())
print(c.info())
print(d.fetch())
print(isinstance(d,Dog))
print(isinstance(d,Animal))
print(isinstance(c,Dog))
print(issubclass(Dog,Animal))
print(type(d) == Animal)
print(type(d) == Dog)
```

```python
class Engine:
    def __init__(self,power):
        self.power = power
        self.running = False
    def start(self):
        self.running = True
        return f"引擎启动({self.power}),马力"
    def stop(self):
        self.running = False
        return f"引擎熄火"
class Radio:
    def __init__(self):
        self.channel = 1
    def play(self):
        return f"播放频道{self.channel}"
class Car:
    def __init__(self,brand,engine):
        self.brand = brand
        self.engine = engine
        self.radio = Radio()
        self.odometer = 0
    def drive(self,km):
        if not self.engine.running:
            return "请先启动引擎"
        self.odometer += km
        return f"行驶了{km}km,总里程{self.odometer}km"
    def info(self):
        return f"{self.brand}|{self.engine.power}马力|里程{self.odometer}km"
engine = Engine(150)
my_car = Car("丰田",engine)
print(my_car.engine.start())
print(my_car.drive(100))
print(my_car.radio.play())
print(my_car.info())
your_car = Car("本田",engine=Engine(200))
print(your_car.info())
```
