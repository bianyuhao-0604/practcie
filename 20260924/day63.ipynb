```python
import numpy as np
a = np.array([1,2,3],dtype=float)
b = np.array([10,20,30],dtype=float)
print(a+b)
print(a*b)
print(a/b)
print(a**2)
print(np.sqrt(a))
print(np.round(np.sqrt(b),2))
```

```python
x = np.linspace(0,2*np.pi,5)
print(np.round(np.sin(x),3))
y = np.array([-1,0,1,np.e,np.e**2])
print(np.abs(y))
print(np.round(np.log(y[y>0]),3))
a = np.array([1,8,3]);b = np.array([4,5,2])
print(np.maximum(a,b))
print(np.max(a))
```

```python
M = np.arange(12).reshape(3,4)
print(M.sum())
print(M.sum(axis=0))
print(M.mean(axis=1))
print(M.mean(axis=1,keepdims=True))
print(M.std(axis=0).round(2))
```

```python
arr = np.arange(6).reshape(2,3)
print(arr + 10)
row = np.array([100,200,300])
print(arr + row)
```

```python
x = np.array([1,2,3])
y = np.array([10,20,30,40])
R = x[:,np.newaxis] + y
print(R)
print(R.shape)
i = np.arange(1,10)
print((i[:,None]*i))
```

```python
arr = np.array([3,-1,7,-5,0,9,2])
print(np.where(arr < 0,0,arr))
print(np.where(arr > 0))
print(arr[np.where(arr > 0)[0]])
conds = [arr < 0,arr == 0,arr > 0 ]
choices = ['neg','zero','pos']
print(np.select(conds,choices,default='unknown'))
```

```python
arr = np.array([3,1,4,1,5,9,2,6])
#print(np.sort(arr))
#print(arr.sort())
idx = np.argsort(arr)
print(idx)
top3_idx = idx[-3:][::-1]
print(top3_idx)
print(arr[top3_idx])
M = np.array([[3,1,2],[6,4,5]])
print(np.sort(M,axis=1))
```

```python
A = np.array([[1,2],[3,4]],dtype=float)
b = np.array([5,11])
print(A@A)
print(A*A)
x = np.linalg.solve(A,b)
print(x)
print(np.linalg.det(A))
print(np.linalg.inv(A).round(3))
print(np.linalg.eigvals(A))
v = np.array([3,4])
print(np.linalg.norm(v))
```

```python
P = np.array([[0,0],[1,2],[3,1],[4,4],[2,5]],dtype=float)
diff = P[:,None,:]-P[None,:,:]
print(P[:,None,:])
print(P[None,:,:])
print(diff)
D = np.sqrt((diff**2).sum(axis=-1))
print(D.round(2))
print(np.allclose(np.diag(D),0))
print(np.allclose(D,D.T))
```
