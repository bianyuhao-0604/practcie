```python
import torch
import torch.nn as nn
class GRU(nn.Module):
    def __init__(self):
        super().__init__()
        self.gru = nn.GRU(input_size=8,hidden_size=40,num_layers=1,batch_first=True)
        self.fc = nn.Linear(40,3)
    def forward(self,x):
        out,h = self.gru(x)
        return self.fc(h[-1])
model = GRU()
model.eval()
with torch.no_grad():
    out = model(torch.randn(4,24,8))
    print(f"'输出形状',{out.shape if isinstance(out,torch.Tensor) else type(out)}")


```

```python
import torch
import torch.nn as nn
class LSTM(nn.Module):
    def __init__(self):
        super().__init__()
        self.lstm = nn.LSTM(input_size=12,hidden_size=48,num_layers=1,batch_first=True)
        self.fc = nn.Linear(48,6)
    def forward(self,x):
        out,(h,c) = self.lstm(x)
        return self.fc(h[-1])
model = LSTM()
model.eval()
with torch.no_grad():
    out = model(torch.randn(4,20,12))
    print('输出形状：', out.shape if isinstance(out, torch.Tensor) else type(out))
    
```
