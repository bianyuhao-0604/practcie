```python
class Rectangle:
    def __init__(self,width,height):
        self.width = width
        self.height = height
    def area(self):
        return self.width * self.height
    def perimeter(self):
        return 2 * (self.width + self.height)
    
r1 = Rectangle(10,20)
print(r1.area())
print(r1.perimeter())
```

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner = owner
        self.balance =balance
    def deposit(self,amount):
        if amount <= 0:
            raise ValueError()
        self.balance += amount
        return self.balance
    def withdraw(self,amount):
        if amount <= 0 or amount > self.balance:
            raise ValueError
        self.balance -= amount
        return self.balance
b1 = BankAccount('张三',1000)
print(b1.deposit(100))
print(b1.withdraw(1000))
```

```python
class Counter:
    def __init__(self):
        self.count = 0
    def increment(self):
        self.count += 1
        return self
    def decrement(self):
        self.count -= 1
        return self
    def get_count(self):
        return self.count
    def reset(self):
        self.count = 0
        return self
c1 = Counter()
c1.increment().increment().increment()
print(c1.get_count())
```

```python
class Temperature:
    def __init__(self,cs):
        self.cs = cs
    def to_fahrenheit(self):
        f = self.cs * 9 / 5 + 32
        return f
    def to_kelvin(self):
        k = self.cs + 273.15
        return k
t1 = Temperature(25)
print(t1.to_fahrenheit())
print(t1.to_kelvin())
        
```

```python
class Book:
    def __init__(self,title,author,pages):
        self.title = title
        self.author = author
        self.pages = pages
    def __str__(self):
        return f"《{self.title}》 - {self.author} ({self.pages}页)"
class BookShelf:
    def __init__(self):
        self.books = []
    def add_book(self,book):
        self.books.append(book)
    def list_books(self):
        for i,book in enumerate(self.books,1):
            print(f'{i}.{book}')
    def total_pages(self):
        return sum(b.pages for b in self.books)

shelf = BookShelf()
shelf.add_book(Book("Python入门", "Guido", 300))
shelf.add_book(Book("算法导论", "CLRS", 1200))
shelf.list_books()
print(f"总页数: {shelf.total_pages()}")



    
```

```python
import math
class Circle:
    def __init__(self,x,y,r):
        self.x = x
        self.y = y
        self.r = r
    def area(self):
        return round(math.pi * self.r ** 2)
    def circumference(self):
        return round(math.pi * self.r * 2)
    def contains_point(self,px,py):
        distance = math.sqrt((px - self.x)**2 + (py-self.y)**2)
        return distance <= self.r

c = Circle(0,0,5)
print(c.area()) 
print(c.circumference())   
print(f"(3,4)在圆内: {c.contains_point(3, 4)}")
```

```python
import random
class Dice:
    def __init__(self,sides=6):
        self.sides = sides
        self.roll_history = []
    def roll(self):
        result = random.randint(1,self.sides)
        self.roll_history.append(result)
        return result
    def roll_multiple(self,times):
        results = [self.roll() for _ in range(times)]
        return results
    
    def statistics(self):
        if not self.roll_history:
            return"还没有投掷记录"
        return {
            "总次数": len(self.roll_history),
            "平均值": round(sum(self.roll_history) / len(self.roll_history), 2),
            "最大值": max(self.roll_history),
            "最小值": min(self.roll_history)
        }

d = Dice(6)
print(d.roll_multiple(7))
print(d.statistics())
```
