```python
file1_lines = {"import os","import sys","print('hello')","x=1"}
file2_lines = {"import os","import json","print('hello')","y=2"}
added_lines = file2_lines - file1_lines
removed_lines = file1_lines - file2_lines
common_lines = file1_lines & file2_lines
print("新增的行: ",added_lines)
print("删除的行: ",removed_lines)
print("共同的行: ",common_lines)
graph = {
    "A": {"B", "C"},
    "B": {"A", "C", "D"},
    "C": {"A", "B"},
    "D": {"B"},
}
common_neighbors = graph["A"] & graph["B"]
print(f"A和B的共同邻居:{common_neighbors}")
only_connected_to_a = graph["A"] - graph["B"]
print(f"只与A相连的节点:{only_connected_to_a}")
article_tags = {
    "文章1": {"Python", "编程", "入门"},
    "文章2": {"Python", "数据", "分析"},
    "文章3": {"Java", "编程", "企业"},
    "文章4": {"Python", "编程", "数据"},
    "文章5": {"JavaScript", "前端", "编程"},
}
search_tags = {"Python", "编程", "数据"}
print("\n===标签匹配===")
matches = []
for article,tags in article_tags.items():
    overlap = len(search_tags & tags)
    if overlap > 0:
        matches.append((article,overlap,search_tags & tags))
matches.sort(key=lambda x:x[1],reverse=True)
for article,count,matched in matches:
    print(f"  {article}:匹配{count}个标签{matched}")
```

```python
from itertools import product,combinations
def power_set(s):
    s = list(s)
    result = []
    for i in range(len(s)+1):
        for combo in combinations(s,i):
            result.append(set(combo))
    return result
s = {1,2,3}
ps = power_set(s)
print(f"集合{s}的幂集(共{len(ps)}个子集):")
for subset in ps:
    print(f"  {subset}")
colors = {"红", "蓝"}
sizes = {"S", "M", "L"}
combos = list(product(colors,sizes))
print(f"\n笛卡尔积(共{len(combos)}种组合):")
for color,size in combos:
    print(f"  {color}-{size}")
```

```python
from collections import Counter,defaultdict
print("="*50)
print("     社交网络分析系统")
print("="*50)
friendships = {
    "小明": {"小红", "小刚", "小丽"},
    "小红": {"小明", "小刚", "小华"},
    "小刚": {"小明", "小红", "小丽", "小陈"},
    "小丽": {"小明", "小刚", "小林"},
    "小华": {"小红", "小陈", "小林"},
    "小陈": {"小刚", "小华"},
    "小林": {"小丽", "小华"},
}
def mutual_friends(network,person1,person2):
    if person1 not in network or person2 not in network:
        return set()
    return network[person1] & network[person2]
def friend_recommendations(network,person):
    if person not in network:
        return set()
    friends = network[person]
    friends_of_friends = set()
    for friend in friends:
        friends_of_friends |= network[friend]
    recommendations = friends_of_friends - friends - {person}
    return recommendations
def degree_centrality(network):
    n = len(network)
    centrality = {}
    for person,friends in network.items():
        centrality[person] = len(friends) / (n-1)
    return centrality
print("\n--- 共同好友 ---")
pairs = [("小明", "小华"), ("小明", "小陈"), ("小红", "小林")]
for p1,p2 in pairs:
    mutual = mutual_friends(friendships,p1,p2)
    print(f"  {p1}和{p2}的共同好友: {mutual}")
print("\n---好友推荐 ---")
for person in ["小明","小红","小陈"]:
    recs = friend_recommendations(friendships,person)
    print(f"   给{person}推荐:{recs}")
print("\n--- 度中心性（社交活跃度）---")
centrality = degree_centrality(friendships)
for person,score in sorted(centrality.items(),key=lambda x:x[1],reverse=True):
    bar = "█" * int(score * 20)
    print(f"  {person}：{score:.2f}  {bar}")
```

```python
print("\n" + "=" * 50)
print("         简易搜索引擎")
print("=" * 50)
documents = {
    "doc1": "Python is a programming language",
    "doc2": "Python is easy to learn",
    "doc3": "Java is a programming language",
    "doc4": "JavaScript is used for web development",
    "doc5": "Python and Java are both popular programming languages",
}
inverted_index = defaultdict(set)
for doc_id,content in documents.items():
    words = content.lower().split()
    for word in words:
        inverted_index[word].add(doc_id)
print("\n倒排索引：")
for word,docs in sorted(inverted_index.items()):
    print(f"  '{word}' → {sorted(docs)}")
def search(query,index):
    words = query.lower().split()
    if not words:
        return set()
    result = index.get(words[0],set().copy())
    for word in words[1:]:
        result &= index.get(word,set())
    return result
queries = ["Python", "programming language", "Python programming", "Java"]
print("\n--- 搜索结果 ---")
for query in queries:
    results = search(query,inverted_index)
    print(f"  搜索 '{query}'：{sorted(results)}")
```

```python
print("\n" + "=" * 50)
print("    集合运算可视化")
print("=" * 50)
def visualize_set_operation(s1,s2,operation_name,result):
    print(f"\n  A = {sorted(s1)}")
    print(f"  B = {sorted(s2)}")
    print(f"  A {operation_name} B = {sorted(result)}")
    only_a = s1 - s2
    both = s1 & s2
    only_b = s2 - s1
    print(f"\n  ┌─────────┐   ┌─────────┐")
    print(f"  │   A      │   │    B    │")
    print(f"  │ {str(sorted(only_a)):^8} │   │ {str(sorted(only_b)):^7} │")
    print(f"  │          ├─{str(sorted(both)):^8}─┤         │")
    print(f"  └─────────┘   └─────────┘")
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}    
visualize_set_operation(a,b,"n交集",a & b)
visualize_set_operation(a, b, "∪（并集）", a | b)
visualize_set_operation(a, b, "-（差集）", a - b)
visualize_set_operation(a, b, "△（对称差集）", a ^ b)
```
