```python
import functools
def simple_cache(func):
    cache = {}
    @functools.wraps(func)
    def wrapper(*args):
        if args not in cache:
            cache[args] = func(*args)
        return cache[args]
    wrapper.cache_clear = lambda: cache.clear()
    return wrapper
@simple_cache
def fib(n):
    if n < 2: return n
    return fib(n-2) + fib(n-1)
print(fib(30))
print(fib.__name__)
fib.cache_clear()
```

```python
def deco_a(func):
    @functools.wraps(func)
    def wrapper(*a,**kw):
        print('A before')
        r = func(*a,**kw)
        print('A after')
        return r
    return wrapper
def deco_b(func):
    @functools.wraps(func)
    def wrapper(*a,**kw):
        print('B before')
        r = func(*a,**kw)
        print('B after')
        return r
    return wrapper
@deco_a
@deco_b
def greet():
    print('Hello!')
greet()
```

```python
def counter():
    print(">>> 开始执行")
    n = 0
    while n < 3:
        print(f">>> 即将 yield {n}")
        yield n
        print(f">>> yield {n} 后继续")
        n += 1
    print(">>> 执行结束")
gen = counter()
print("--- 第一次 next ---")
print(next(gen))
print("--- 第二次 next ---")
print(next(gen))
print("--- 第三次 next ---")
print(next(gen))
print("--- 第四次 next ---")
try:
    next(gen)
except StopIteration:
    print("生成器已耗尽")
```

```python
def simulate_large_file(num_lines):
    for i in range(num_lines):
        yield f"line-{i}: some log data here\n"
def process_log(file_gen,keyword):
    count = 0
    for line in file_gen:
        if keyword in line:
            count += 1
            yield line.strip()
    return count
log_gen = simulate_large_file(1_000_000)
filtered = process_log(log_gen,"line-9999")
for line in filtered:
    print(line)
```

```python
def accumulator():
    total = 0
    while True:
        value = yield total
        if value is None:
            break
        total += value
gen = accumulator()
next(gen)
print(gen.send(10))
print(gen.send(20))
print(gen.send(30))
gen.close()
```
