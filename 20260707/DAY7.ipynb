```python
class BankAccount:
    def __init__(self,owner:str,initial_balance:float = 0):
        self.owner = owner
        self._balance = initial_balance
    @property
    def balance(self) -> float:
            return self._balance
        
    def deposit(self,amount:float) -> None:
            if amount <= 0:
                raise ValueError("存款金额必须大于0")
            self._balance += amount
            print(f"✅ 存入 {amount}，当前余额: {self._balance}")
    def withdraw(self,amount:float) -> None:
            if amount <= 0:
                raise ValueError("取款金额必须大于0")
            if amount > self._balance:
                raise ValueError(f"余额不足！当前余额: {self._balance}")
            self._balance -= amount
            print(f"✅ 取出 {amount}，当前余额: {self._balance}")
    def __str__(self) -> str:
            return f"BankAccount({self.owner}, 余额: {self._balance})"
account = BankAccount("张三", 1000)
print(account.owner)
account.deposit(500)
account.withdraw(200)
print(f"当前余额: {account.balance}")

   
```

```python
class Temperature:
    def __init__(self,celsius:float = 0):
        self._celsius = celsius
    @property
    def celsius(self) -> float:
        return self._celsius

    @celsius.setter
    def celsius(self,value:float) -> None:
        self._celsius = value
    
    @property
    def fahrenheit(self) -> float:
        return self._celsius * 9 /5 + 32
    
    @fahrenheit.setter
    def fahrenheit(self,value:float) -> None:
        self._celsius = (value - 32) * 5 / 9
    
    @property
    def kelvin(self) -> float:
        return self._celsius + 273.15
    
    def __repr__(self) -> str:
        return f'Temperature({self._celsius}℃  {self.fahrenheit:.1f}℉ / {self.kelvin:.1f}K)'
temp = Temperature(25)
print(temp)


    
```

```python
class Rectangle:
    def __init__(self,width:float,height:float):
        self._width = width
        self._height = height
    @property
    def width(self) -> float:
        return self._width
    
    @width.setter
    def width(self,value:float) -> None:
        self._width = value
    
    @property
    def height(self) -> float:
        return self._height

    @height.setter
    def height(self,value:float) -> float:
        self._height = value
    
class Square:
    def __init__(self,side:float):
        self._side = side

    @property
    def side(self) -> float:
        return self._side
    
    @side.setter
    def side(self,value:float) -> None:
        self._side = value
    
    @property
    def area(self) -> float:
        return self._side ** 2




```

```python
from dataclasses import dataclass
from typing import List

@dataclass
class Course:
    """课程类"""
    name: str
    credit: float
    score: float  # 百分制
    
    @property
    def grade_point(self) -> float:
        """转换为绩点（4.0制）"""
        if self.score >= 90:
            return 4.0
        elif self.score >= 85:
            return 3.7
        elif self.score >= 82:
            return 3.3
        elif self.score >= 78:
            return 3.0
        elif self.score >= 75:
            return 2.7
        elif self.score >= 72:
            return 2.3
        elif self.score >= 68:
            return 2.0
        elif self.score >= 64:
            return 1.5
        elif self.score >= 60:
            return 1.0
        else:
            return 0.0

class Student:
    def __init__(self, name: str, student_id: str):
        self.name = name
        self.student_id = student_id
        self._courses: List[Course] = []
    
    def add_course(self, course: Course) -> None:
        """添加课程"""
        self._courses.append(course)
        print(f"✅ {self.name} 选修了 {course.name}")
    
    @property
    def courses(self) -> List[Course]:
        """只读访问课程列表"""
        return self._courses.copy()  # 返回副本，防止外部修改
    
    @property
    def gpa(self) -> float:
        """计算加权平均分（GPA）"""
        if not self._courses:
            return 0.0
        total_credits = sum(c.credit for c in self._courses)
        weighted_sum = sum(c.grade_point * c.credit for c in self._courses)
        return weighted_sum / total_credits
    
    @property
    def total_credits(self) -> float:
        """总学分"""
        return sum(c.credit for c in self._courses)
    
    def __str__(self) -> str:
        return f"Student({self.name}, GPA: {self.gpa:.2f}, 学分: {self.total_credits})"

# 测试
student = Student("李四", "2024001")
student.add_course(Course("高等数学", 4, 92))
student.add_course(Course("线性代数", 3, 85))
student.add_course(Course("Python编程", 2, 95))

print(student)  # GPA: 3.83, 学分: 9
print(f"详细成绩:")
for course in student.courses:
    print(f"  {course.name}: {course.score}分 (绩点 {course.grade_point})")
```

```python
from typing import List, Any

class Stack:
    """栈数据结构（后进先出）"""
    
    def __init__(self):
        self._items: List[Any] = []
    
    def push(self, item: Any) -> None:
        """入栈"""
        self._items.append(item)
    
    def pop(self) -> Any:
        """出栈"""
        if self.is_empty():
            raise IndexError("栈为空！")
        return self._items.pop()
    
    def peek(self) -> Any:
        """查看栈顶元素（不出栈）"""
        if self.is_empty():
            raise IndexError("栈为空！")
        return self._items[-1]
    
    def is_empty(self) -> bool:
        """是否为空"""
        return len(self._items) == 0
    
    # ========== 魔术方法 ==========
    
    def __len__(self) -> int:
        """支持 len(stack)"""
        return len(self._items)
    
    def __contains__(self, item: Any) -> bool:
        """支持 item in stack"""
        return item in self._items
    
    def __iter__(self):
        """支持 for item in stack（从栈顶到栈底）"""
        return reversed(self._items)
    
    def __str__(self) -> str:
        """友好的字符串表示"""
        if self.is_empty():
            return "Stack(空)"
        items_str = " <- ".join(str(item) for item in reversed(self._items))
        return f"Stack(栈顶: {items_str})"
    
    def __repr__(self) -> str:
        """开发者友好的表示"""
        return f"Stack({self._items})"

# 测试
stack = Stack()
stack.push("A")
stack.push("B")
stack.push("C")

print(stack)  # Stack(栈顶: C <- B <- A)
print(f"大小: {len(stack)}")  # 3
print(f"包含 B: {'B' in stack}")  # True
print(f"栈顶: {stack.peek()}")  # C

print("遍历（从栈顶到栈底）:")
for item in stack:
    print(f"  {item}")

print(f"出栈: {stack.pop()}")  # C
print(stack)  # Stack(栈顶: B <- A)
```
