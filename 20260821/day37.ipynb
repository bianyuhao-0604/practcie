```python
def detect_encoding(data:bytes,encodings=None) -> str | None:
    if encodings is None:
        encodings = ['utf-8','gbk','latin-1','utf-16']
    for enc in encodings:
        try:
            data.decode(enc)
            return enc
        except (UnicodeDecodeError,LookupError):
            continue
    return None
test_cases = [
    b'Hello World',
    '你好世界'.encode('gbk'),
    'こんにちは'.encode('utf-8'),
    b'\xff\xfe\x00\x00',
]
for tc in test_cases:
    result = detect_encoding(tc)
    print(f"{tc!r:30s} → {result}")
```

```python
def make_counter(start=0):
    count = start
    def counter():
        nonlocal count
        count += 1
        return count
    return counter
c = make_counter(10)
print(c())
print(c())
print(c.__closure__[0].cell_contents)
```
