```python
num = [1,2,3,4,5]
a = [x ** 2 for x in num]
print(a)
```

```python
nums = range(1,22)
a = [x ** 3 for x in nums if x % 2 == 0]
print(a)
```

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
flat = [val for row in matrix for val in row]
print(flat)
```

```python
words = ['hello','world','python']
a = {x:len(x) for x in words}
print(a)
```

```python
scores = {'Alice':92,'Bob':85,'Charlie':78}
for name ,score in scores.items():
    print(f'{name},{score}分')
```

```python
fruits = ['apple','banana','cherry']
for i,fruit in enumerate(fruits,start = 1):
    print(f'{i}.{fruit}')
```

```python
names = ['Alice','Bob']
ages = [25,30]
for name,age in zip(names,ages):
    print(f'{name} is {age} years old.')
```

```python
keyword = 'py2026'
a = input('请输入密码:')
count = 3
while count > 0:
    if a == keyword:
        print('密码正确')
    else:
        count -= 1
        print(f'密码错误,剩余次数为{count}')
    break


```

```python
class_a = ['Tom','Jerry','Alice']
class_b = ['Alice','Bob','Tom']
c  = set(class_a) & set(class_b)
d = len(set(class_a) | set(class_b))
print(c,d)
```

```python
from collections import Counter 
text = "to be or not to be that is the question to be"
words = text.lower().split()
top3 = Counter(words).most_common(3)
for word ,count in top3:
    print(f"'{word}':{count}次")
```
