```python
import re
text = '我的电话是138-1234-5678,备用号139-8765-4321'
#re.search 从头查找第一个匹配的
result = re.search(r'\d{3}-\d{4}-\d{4}',text)
print(result.group())

#re.match 只看开头是否匹配
result = re.match(r'\d+',text)
print(result)
result = re.match(r'我的',text)
print(result.group())

#re.findall 找所有
result = re.findall(r'\d{3}-\d{4}-\d{4}',text)
print(result)
```

```python
import re
text = 'Hello World! 你好2024,价格$99.9元。邮箱: test@example.com'
print(re.findall(r'\d+',text))
print(re.findall(r'\w+',text))
print(re.findall(r'\s+',text))
print(re.findall(r'H.llo',text))
print(re.findall(r'\D+',text))
print(re.findall(r'\S','abc123def'))
print(re.findall(r'^Hello',text))
print(re.findall(r'com$',text))
print(re.findall(r'[example]+',text))
print(re.findall(r'[^example]+',text))
```

```python
import re
text = "今天是 2024-12-25,明天是 2024-12-26"
pattern = r'(\d{4})-(\d{2})-(\d{2})'

match = re.search(pattern,text)
print(match.group(0))
print(match.group(1))
print(match.group(2))
print(match.group(3))

results = re.findall(pattern,text)
print(results)

pattern = r'(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})'
match = re.search(pattern,text)
print(match.group())
print(match.group(1))
print(match.group(2))
print(match.group(3))

text = 'cat bat hat'
print(re.findall(r'(c|b|h)at',text))
print(re.findall(r'(?:c|b|h)at',text))
```

```python
import re
html = '<div>hello</div><div>world</div>'
result = re.findall(r'<div>(.*)</div>',html)
print(result)

result = re.findall(r'<div>(.*?)</div>',html)
print(result)

text = 'name="Alice" age="30"'
print(re.findall(r'\"(.*)\"',text))
print(re.findall(r'\"(.*?)\"',text))
```

```python
import re

phone_pattern = r'1[3-9]\d{9}'
text = '联系我:13812345678或15987654321,座机010-12345678'
print(re.findall(phone_pattern,text))

email_pattern = r'[\w.+-]+@[\w-]+\.[\w.]+'
text = '发邮件到alice@example.com 和 bob.smith+tag@company.co.jp'
print(re.findall(email_pattern,text))

url_pattern = r'https?://[\w./\-?=&%]+'
text = "访问 https://www.baidu.com/s?wd=python 或 http://example.com"
print(re.findall(url_pattern,text))

id_pattern = r'\d{17}[\dXx]'
text = '身份证号:110101199001011234 和 44030119851215000X'
print(f"身份证: {re.findall(id_pattern, text)}")

chinese_pattern = r'[\u4e00-\u9fff]+'
text = "Hello你好World世界Python编程"
print(f"中文: {re.findall(chinese_pattern, text)}")
```

```python
import re 
text = '张三 13812345678,李四15987654321'
result = re.sub(r'(1\d{2})\d{4}(\d{4})',r'\1****\2',text)
print(result)

def mask_phone(match):
    phone = match.group()
    return phone[:3] + '****' + phone[7:]

result = re.sub(r'1[3-9]\d{9}',mask_phone,text)
print(result)

messy = 'hello  world  python   正则'
clean = re.sub(r'\s+','',messy).strip()
print(clean)

def camel_to_snake(name):
    return re.sub(r'([A-Z])',r'_\1',name).lower().lstrip('_')
print(camel_to_snake('getUserName'))
print(camel_to_snake('HTTPServer'))

template = '你好，{name}！你的订单 {order_id} 已发货。'
data = {"name": "小明", "order_id": "2024001"}
result = re.sub(r'\{(\w+)\}', lambda m: data.get(m.group(1), m.group(0)), template)
print(result)
```
