```python
from enum import Enum
class OrderStatus(Enum):
    PENDING = "pending"
    PAID = "paid"
    SHIPPED = "shipped"
    DELIVERED = "delivered"
    CANCELLED = "cancelled"
current_status = OrderStatus.PAID
print(current_status == OrderStatus.PAID)
print(current_status == 'paid')
print(current_status == OrderStatus.SHIPPED)
print(current_status.name)
print(current_status.value)
status = OrderStatus('shipped')
print(status)
```

```python
from enum import Enum
class Permission(Enum):
    READ = 1
    WRITE = 2
    EXECUTE = 4
    ADMIN = 8
for perm in Permission:
    print(f'{perm.name:10s} == {perm.value}')
def is_valid_permission(value):
    try:
        Permission(value)
        return True
    except ValueError:
        return False
print(is_valid_permission(3))
print(is_valid_permission(8))
perm = Permission['READ']
print(perm)
all_perms = list(Permission)
print(len(all_perms))
```

```python
from enum import IntEnum,Flag,auto
class FilePerm(IntEnum):
    READ = 4
    WRITE = 2
    EXECUTE = 1
user_perm = FilePerm.READ | FilePerm.WRITE
print(user_perm)
print(bool(user_perm & FilePerm.READ))
print(bool(user_perm & FilePerm.EXECUTE))

print(FilePerm.READ == 4)

class AccessFlag(Flag):
    READ = auto()
    WRITE = auto()
    EXECUTE = auto()
    DELETE = auto()
flags = AccessFlag.READ | AccessFlag.WRITE
print(flags)
print(AccessFlag.READ in flags)
print(flags == 3)
```

```python
from typing import NamedTuple
class Point(NamedTuple):
    lat: float
    lng: float
    name: str = "未知地点"

p1 = Point(122.231,92.231,'北京天安门')
p2 = (122.213,213.231)
print(p1,p2)
print(p1 == p2)
print(p1.name,p1.lat,p1.lng)
#p1.lng = 122.124 不可改变，报错
p3 = p1._replace(name='故宫')
print(p3)
print(p1.name,p3.name)
```

```python
import sys
from typing import NamedTuple
class RegularPoint:
    def __init__(self,x,y,z):
        self.x = x
        self.y = y
        self.z = z

class NTPoint(NamedTuple):
    x: float
    y:float
    z:float
rp = RegularPoint(1.0,2.0,3.0)
np_ = NTPoint(1.0,2.0,3.0)
reg_size = sys.getsizeof(rp) + sys.getsizeof(rp.__dict__)
nt_size = sys.getsizeof(np_)

print(f'普通类 {reg_size} bytes')
print(f'NamedTuple {nt_size} bytes')
print(f'节省:{(1-nt_size/reg_size) * 100:.0f}%')
```

```python
from dataclasses import dataclass
@dataclass
class Product:
    name:str
    price:float
    stock:int = 0
p1 = Product('机械键盘',999.0,1000)
p2 = Product('固态硬盘',1499.0,2000)
print(p1 == p2)
print(p1.name,p1.price,p1.stock)
print(p1)
```

```python
from dataclasses import dataclass,field
@dataclass(frozen=True)
class Color:
    r:int
    g:int
    b:int

c = Color(255,128,0)
print(c)
#c.r = 129
color_map = {Color(255,0,0):'红色',Color(0,255,0):'绿色'}
print(color_map)

@dataclass(order=True)
class Version:
    major:int
    minor:int
    patch:int
v1 = Version(1,2,3)
v2 = Version(1,3,0)
print(v1 > v2)

@dataclass
class Team:
    name:str
    members:list = field(default_factory=list)
    metadata:dict = field(default_factory=dict)
t1 = Team('Alpha')
t2 = Team('Beta')
t1.members.append('Alice')
print(t1)
print(t1.members,t1.name,t2.members)
```

```python
from dataclasses import dataclass,field
from datetime import datetime
@dataclass
class Task:
    title:str
    priority:int = 0
    created_at:datetime = field(default_factory=datetime.now)
    done:bool = False
    def __post_init__(self):
        if self.priority < 0 or self.priority > 10:
            raise ValueError(f'优先级必须在 0-10 之间，收到: {self.priority}')
        if len(self.title) > 50:
            self.title = self.title[:47] + '...'
    def complete(self):
        self.done =True
        print(f"✅ '{self.title}' 已完成!")
    
    @property
    def urgency_label(self):
        if self.priority >= 8: return "🔴 紧急"
        if self.priority >= 5: return "🟡 普通"
        return "🟢 低优"
task = Task('学习 Python dataclass',priority=7)
print(task)        
print(task.urgency_label)
task.complete()
print(task.done)
```

```python
from enum import Enum
from typing import NamedTuple
from dataclasses import dataclass
class Suit(Enum):
    CLUBS = "♣"
    DIAMONDS = "♦"
    HEARTS = "♥"
    SPADES = "♠"
class Rank(Enum):
    TWO = 2; THREE = 3; FOUR = 4; FIVE = 5
    SIX = 6; SEVEN = 7; EIGHT = 8; NINE = 9
    TEN = 10; JACK = 11; QUEEN = 12; KING = 13; ACE = 14
class CardNT(NamedTuple):
    rank:Rank
    suit:Suit
    def __str__(self):
        return f'{self.suit.value}{self.rank.name}'
card_nt = CardNT(Rank.ACE,Suit.SPADES)
print(f"NamedTuple 牌: {card_nt}")

@dataclass(order=True)
class CardDC:
    rank:Rank
    suit: Suit
    def __str__(self):
        return f"{self.suit.value}{self.rank.name}"
card_dc = CardDC(Rank.KING,Suit.HEARTS)
print(f"Dataclass 牌: {card_dc}")
print(f"Ace > King? {CardNT(Rank.ACE, Suit.SPADES).rank.value > card_dc.rank.value}")
```
