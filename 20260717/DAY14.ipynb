```python
from datetime import datetime
class Product:
    def __init__(self,name,price,stock):
        self.name = name
        self.price = price
        self.stock = stock
        self.created_at = datetime.now()
    def __str__(self):
        return f'{self.name} - ¥{self.price:.2f} (库存: {self.stock})'
    def __repr__(self):
        return f'Product:name={self.name},price={self.price},stock={self.stock}'
    
p = Product('机械键盘',350,2000)
print(p)
print(repr(p))

products = [p,Product('鼠标',99.0,100)]
print(products)
```

```python
import math
class Vector:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __repr__(self):
        return f'Vector({self.x},{self.y})'
    def __add__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x + other.x,self.y + other.y)
        return NotImplemented
    def __sub__(self,other):
        if isinstance(other,Vector):
            return Vector(self.x - other.x,self.y - other.y)
        return NotImplemented
    def __mul__(self,scalar):
        if isinstance(scalar,(int,float)):
            return Vector(self.x * scalar,self.y * scalar)
        return NotImplemented
    def __rmul__(self,scalar):
        return self.__mul__(scalar)
    def __matmul__(self,other):
        if isinstance(other,Vector):
            return self.x * other.x + self.y * other.y
        return NotImplemented
    def __abs__(self):
        return math.hypot(self.x,self.y)
    def __neg__(self):
        return Vector(-self.x,-self.y)

v1 = Vector(3,4)
v2 = Vector(1,2)
print(f'v1 + v2 = {v1+v2}')
print(f'v1 - v2 = {v1-v2}')
print(v1 @ v2)
print(abs(v1))    
print(-v1)
```

```python
from functools import total_ordering
EXCHANGE_RATES = {
    'USD':1.0,
    'CNY':7.25,
    'EUR':0.92,
    'JPY':157.5,
}

@total_ordering
class Money:
    def __init__(self,amount,currency='USD'):
        if currency not in EXCHANGE_RATES:
            return ValueError(f"不支持的货币: {currency}")
        self.amount = amount
        self.currency = currency
    def to_usd(self):
        return self.amount / EXCHANGE_RATES[self.currency]
    def __eq__(self,other):
        if not isinstance(other,Money):
            return NotImplemented
        return abs(self.to_usd() - other.to_usd()) < 1e-9
    def __lt__(self,other):
        if not isinstance(other,Money):
            return NotImplemented
        return self.to_usd() < other.to_usd()
    
    def __repr__(self):
        return f"Money({self.amount:.2f}, '{self.currency}')"
    def __add__(self,other):
        if isinstance(other,Money):
            total_usd = self.to_usd() + other.to_usd()
            new_amount = total_usd * EXCHANGE_RATES[self.currency]
            return Money(new_amount,self.currency)
        return NotImplemented
    
m1 = Money(100,'USD')
m2 = Money(725,'CNY')
m3 = Money(50,'EUR')

print(m1 == m2)
print(m3 < m1)
print(m1 >= m3)

wallet = [Money(1000, "JPY"), Money(10, "USD"), Money(50, "CNY")]
print(sorted(wallet))
```

```python
class Coordinate:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __eq__(self,other):
        if not isinstance(other,Coordinate):
            return NotImplemented
        return self.x == other.x and self.y == other.y
    def __hash__(self):
        return hash((self.x,self.y))
    def __repr__(self):
        return f"Coordinate({self.x}, {self.y})"
c1 = Coordinate(0,0)
c2 = Coordinate(3,4)
c3 = Coordinate(3,4)
print(c1 == c2)  
print(c1 == c3)  
print(c2 == c3)

points = {c1,c2,c3}
print(points)
```

```python
class SortedDict:
    def __init__(self,data=None):
        self._store = {}
        if data:
            for k,v in data.items():
                self._store[k] = v
    def __setitem__(self,key,value):
        self._store[key] = value
    def __getitem__(self,key):
        try:
            return self._store[key]
        except KeyError:
            raise KeyError(f"键 '{key}' 不存在") from None
    def __delitem__(self,key):
        del self._store[key]
    def __contains__(self,key):
        return key in self._store
    def __len__(self):
        return len(self._store)
    def __iter__(self):
        return iter(sorted(self._store.keys()))
    def __repr__(self):
        items = [(k,self._store[k]) for k in sorted(self._store)]
        return f"SortedDict({dict(items)})"
    def items(self):
        for k in sorted(self._store):
            yield k,self._store[k]

```

```python
class Student:
    def __init__(self,name,age,grade):
        self.name = name
        self.age = age
        self.grade = grade
    def __str__(self):
        return f'{self.name},{self.age}岁,{self.grade}年级'
    def __repr__(self):
        return f'name={self.name},age={self.age},grade={self.grade}'
s1 = Student('张三',16,8)

print(s1)
print(repr(s1))
```

```python
class TodoList:
    def __init__(self):
        self.tasks = []
    def add(self,task):
        self.tasks.append(task)
    def __len__(self):
        return len(self.tasks)
    def done(self,task):
        if task in self.tasks:
            self.tasks.remove(task)
            print(f'{self.task}已完成')
    def __str__(self):
        return f'还有以下任务待完成{self.tasks}'
m1 = TodoList()
m1.add("学 Python 魔术方法")
print(m1)

```

```python
class Playlist:
    def __init__(self,name):
        self.name = name
        self.songs = []
    def add(self,song):
        if song not in self.songs:
            self.songs.append(song)
    def __getitem__(self,index):
        return self.songs[index]
    def __len__(self):
        return len(self.songs)
    def __str__(self):
        return f'{self.name},{len(self)}'



pl = Playlist("我的最爱")
pl.add("晴天 - 周杰伦")
pl.add("稻香 - 周杰伦")
pl.add("起风了 - 买辣椒也用券")
pl.add("光年之外 - 邓紫棋")
print(pl[0])
print(pl[2])
print(pl)
```

```python
class UserGroup:
    def __init__(self,group_name):
        self.group_name = group_name
        self._group = []
    def add(self,username):
        self._group.append(username)
    def __contains__(self,username):
        return username.lower() in self._group
    def __len__(self):
        return len(self._group)



group = UserGroup("Python学习群")
group.add("alice")
group.add("Bob")
group.add("小明")

print(len(group))
print('AlicE' in group)
```
