```python
class Product:
    def __init__(self,name,price,quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
    def __str__(self):
        return f"商品: {self.name}, 价格: ¥{self.price}, 库存: {self.quantity}"
    def __repr__(self):
        return f"Product('{self.name}', {self.price}, {self.quantity})"
p = Product('iphone',7999,100)
print(p)
print(repr(p))
    
    
```

```python
class Date:
    def __init__(self,year,month,day):
        self.year=year
        self.month=month
        self.day=day
    
    @classmethod
    def from_string(cls,date_str):
        year,month,day = map(int,date_str.split('-'))
        return cls(year,month,day)
    
    def __str__(self):
        return f'{self.year}年{self.month}月{self.day}日'
d = Date.from_string("2026-07-13")
print(d)

```

```python
class Validator:
    @staticmethod
    def is_valid_email(email):
        return '@' in email and '.' in email.split('@')[-1]
    @staticmethod
    def is_valid_age(age):
        return isinstance(age,int) and 0 < age < 150
    @staticmethod
    def is_valid_phone(phone):
        return len(phone) == 11 and phone.isdigit()
print(Validator.is_valid_email('test@qq.com'))
print(Validator.is_valid_age(28))
print(Validator.is_valid_phone('13812345678'))
```

```python
class Playlist:
    def __init__(self,name):
        self.name = name
        self.songs = []
    def add_song(self,song):
        return self.songs.append(song)
    def __len__(self):
        return len(self.songs)
    def __genitem__(self,index):
        return self.songs[index]
    def __str__(self):
        return f'播放列表[{self.name}]: {len(self)}首歌'
q = Playlist('我的最爱')
q.add_song('晴天')
q.add_song('七里香')
q.add_song('稻香')
print(len(q))
    
```

```python
class WordList:
    def __init__(self,words):
        self.words = words
    def __contains__(self,word):
        return word.lower() in [w.lower() for w in self.words]
w1 = WordList(['Python','Java','C++'])
print('python' in w1)
```

```python
class Vector2D:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __add__(self,other):
        if isinstance(other,Vector2D):
            return Vector2D(self.x + other.x,self.y + other.y)
        raise TypeError("只能与另一个Vector2D相加")
    def __sub__(self,other):
        if isinstance(other,Vector2D):
            return Vector2D(self.x - other.x,self.y - other.y)
        raise TypeError("只能与另一个Vector2D相减")
    def __mul__(self,scalar):
        if isinstance(scalar,(int,float)):
            return Vector2D(self.x * scalar,self.y * scalar)
        raise TypeError("只能与数字相乘")
    def __repr__(self):
        return f'Vector2D({self.x}, {self.y})'
v1 = Vector2D(2,3)
v2 = Vector2D(5,1)
print(v1 + v2)
print(v1 - v2)
print(v1 * 2)
v3 = (v1 + v2) * 2
print(v3)
```

```python
from functools import total_ordering
@total_ordering
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def __eq__(self,other):
        if not isinstance(other,Student):
            return NotImplemented
        return self.score == other.score
    def __lt__(self,other):
        if not isinstance(other,Student):
            return NotImplemented
        return self.score < other.score
    def __repr__(self):
        return f'{self.name}({self.score}分)'
s1 = Student('zhangsan',80)
s2 = Student('lisi',99)
print(s1 > s2)
print(s1 == s2)
print(s1 < s2)

students = [Student('a',70),Student('b',95),Student('c',88),Student('d',60)]
print("升序:", sorted(students))
print("降序:", sorted(students, reverse=True))
    
```

```python
class CallCounter:
    def __init__(self,func):
        self.func = func
        self.count = 0
    def __call__(self,*args,**kwargs):
        self.count += 1
        print(f'[{self.func.__name__}] 第 {self.count} 次被调用')
        return self.func(*args,**kwargs)

@CallCounter
def greet(name):
    return f'Hello, {name}!'

print(callable(greet))

print(greet('alice'))
print(greet('bob'))
print(greet('charlie'))

print(f'总计调用次数: {greet.count}')
```

```python
class DatabaseConnection:
    _instance = None
    _is_connected = False

    def __new__(cls,*args,**kwargs):
        if cls._instance is None:
            print(">> 创建新的数据库连接实例...")
            cls._instance = super().__new__(cls)
        else:
            print(">> 复用已有的数据库连接实例...")
        return cls._instance
    
    def __init__(self,host='localhost',port=3306):
        if not self._is_connected:
            self.host = host
            self.port = port
            self.is_connected = True
            print(f">> 数据库已连接到 {self.host}:{self.port}")
    
    def query(self,sql):
        return f'执行SQL: {sql}'

db1 = DatabaseConnection('192.168.1.100',3306)
db2 = DatabaseConnection("10.0.0.1", 5432)

print(db1 is db2)
print(db2.host)
```

```python
class FileWriter:
    def __init__(self,filename):
        self.filename = filename
        self.file = None
    
    def __enter__(self):
        print(f"--> 准备打开文件: {self.filename}")
        self.file = open(self.filename,'w',encoding='utf-8')
        return self
    def __exit__(self,exc_type,exc_val,exc_tb):
        print(f"<-- 准备关闭文件: {self.filename}")
        if self.file:
            self.file.close()

        if exc_type is not None:
            print(f"!! 拦截到异常: {exc_type.__name__} - {exc_val}")
            return False
        
    def write(self,content):
        self.file.write(content + '\n')

print("=== 测试 1: 正常流程 ===")
with FileWriter("test_log.txt") as fw:
    fw.write("第一行日志")
    fw.write("第二行日志")

print("\n=== 测试 2: 异常流程 ===")
try:
    with FileWriter("test_log.txt") as fw:
        fw.write("写入成功")
        raise ValueError("模拟一个写入过程中的网络错误！")
except ValueError as e:
    print(f"外部捕获到异常: {e}")
```
