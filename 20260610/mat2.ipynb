```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei']#调用字体
plt.rcParams['axes.unicode_minus'] = False
x = np.linspace(0,10,100)#生成0到10以内1000个随机数
y1 = np.sin(x)
y2 = np.cos(x)
categories = ['A','B','C','D','E']
values = [23,45,56,78,32]
data = np.random.randn(1000)
fig = plt.figure(figsize = (20,24))#生成20*24的画布
gs = fig.add_gridspec(6,4)
ax3 = fig.add_subplot(gs[1,0:2])
bars = ax3.bar(categories,values,color = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd'])
ax3.set_title('3. 柱状图 (Bar Chart)')#设置标题
ax3.set_xlabel('类别')#x轴标签为‘类别’
ax3.set_ylabel('数值')#y轴标签为‘数值’
ax3.bar_label(bars,padding = 3)#为每个柱子添加数值标签
```

```python
import matplotlib.pyplot as plt
import numpy as np
plt.rcParams['font.sans-serif'] = ['SimHei']#调用字体
plt.rcParams['axes.unicode_minus'] = False
x = np.linspace(0,10,100)#生成0到10以内1000个随机数
y1 = np.sin(x)
y2 = np.cos(x)
categories = ['A','B','C','D','E']
values = [23,45,56,78,32]
data = np.random.randn(1000)
fig = plt.figure(figsize = (20,24))#生成20*24的画布
gs = fig.add_gridspec(6,4)
ax5 = fig.add_subplot(gs[2,:])
group1_values = [10,20,15,30,25]
group2_values = [15, 25, 20, 20, 35]
ax5.bar(categories,group1_values,label = '组1',color = 'skyblue')
ax5.bar(categories,group2_values,bottom = group1_values,label = '组2',color = 'salmon')
ax5.set_title('5. 堆叠柱状图 (Stacked Bar)')
ax5.set_xlabel('类别')
ax5.set_ylabel('数值')
ax5.legend()
```
