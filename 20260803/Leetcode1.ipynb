```python
class Solution:
    def twoSum(self,nums:list[int],target:int) -> list[int]:
        dict1 = {}
        for index,item in enumerate(nums):
            if target - item in dict1:
                return [dict1[target-item],index]
            else:
                dict1[item] = index
        return []
s = Solution()
print(s.twoSum([1,2,3,4,5,6],8))
```

```python
class Solution:
    def isPalindrome(self,x:int) -> bool:
        return str(x) == str(x)[::-1]
q = Solution()
print(q.isPalindrome(-3123))
print(q.isPalindrome(12321))
```
