```python
numbers = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
print(f"\n原始列表:{numbers}")
unique_set = list(set(numbers))
print(f"set去重(无序):{unique_set}")
unique_ordered = []
for num in numbers:
    if num not in unique_ordered:
        unique_ordered.append(num)
print(f"手动去重(保序):{unique_ordered}")
list_a = [1, 2, 3]
list_b = [4, 5, 6]
merged = list_a + list_b
print(f"\n合并(+):{merged}")
list_a.extend(list_b)
print(f"合并(extend):{list_a}")
repeated = [0] * 5
print(f"重复(*){repeated}")
nums = [1, 2, 3, 4, 5, 6, 7, 8]
mid = len(nums) // 2
first_half = nums[:mid]
second_half = nums[mid:]
print(f"前半部分{first_half}")
print(f"后半部分{second_half}")
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
print("\n=== 矩阵操作 ===")
print(f"矩阵{matrix}")
print(f"第2行{matrix[1]}")
print(f"第2行第3列{matrix[1][2]}")
print("逐行打印：")
for row in matrix:
    for item in row:
        print(item,end="\t")
    print()
matrix[0][0] = 99
print(f"修改后第1行{matrix[0]}")
```

```python
class shopping_list:
    def __init__(self):
        self.items = []
    def add_item(self,name,price):
        if price < 0:
            raise ValueError("价格不能为负数")
        self.items.append([name,float(price)])
        print(f"已添加:{name}￥{price:.2f}")
    def remove_item(self,name):
        for i,(item_name,_) in enumerate(self.items):
            if item_name == name:
                removed = self.items.pop(i)
                print(f"已删除:{removed[0]}￥{removed[1]:.2f}")
                return True
        print(f"未找到商品:{name}")
        return False
    def update_price(self,name,new_price):
        if new_price < 0:
            raise ValueError("价格不能为负数")
        for item in self.items:
            if item[0] == name:
                old_price = item[1]
                item[1] = float(new_price)
                print(f"已修改 {name} 价格：¥{old_price:.2f} → ¥{new_price:.2f}")
                return True
        print(f"未找到商品:{name}")
        return False
    def total_price(self):
        return sum(item[1] for item in self.items)
    def print_receipt(self):
        print("=" * 34)
        print("购物小票".center(30))
        print("=" * 34)
        if not self.items:
            print("(清单为空)")
        else:
            for i,(name,price) in enumerate(self.items,1):
                print(f"{i:>2}.{name:<15}￥{price:>8.2f}")
        print("-" * 34)
        print(f"总价:￥{self.total_price():.2f}")
        print("=" * 34)
if __name__ == '__main__':
    shop = shopping_list()
    shop.add_item("苹果",5.5)
    shop.add_item("牛奶",12.0)
    shop.add_item("面包",8.0)
    shop.print_receipt()
    shop.update_price("牛奶",10.5)
    shop.remove_item("面包")
    shop.print_receipt()
```
