```python
import copy
a = [1,[2,3],4]
b = a
c = a.copy()
d = copy.deepcopy(a)
a[1].append(99)
a[0] = 100
print('b:',b)
print('c:',c)
print('d:',d)
```

```python
def add_tag(item,tags=None):
    if tags is None:
        tags = []
    tags.append(item)
    return tags
print(add_tag("python"))
print(add_tag("java"))
print(add_tag("go"))
```

```python
original = {"db": {"host": "localhost", "port": 3306}, "debug": True}
config_copy = original.copy()
config_copy['debug'] = False
config_copy['db']['port'] = 5432
print('original:',original)
print('copy: ',config_copy)
```

```python
import copy
DEFAULT_CONFIG = {
    "database": {"host": "localhost", "port": 3306, "pool_size": 5},
    "cache": {"backend": "redis", "ttl": 300},
    "features": ["logging", "metrics"],
}
def get_config():
    return copy.deepcopy(DEFAULT_CONFIG)
module_a_config = get_config()
module_b_config = get_config()
module_a_config['database']['port'] = 5432
module_a_config['features'].append('tracing')
module_b_config['cache']['ttl'] = 600
print('Module A:',module_a_config)
print('Module B:',module_b_config)
print('Default:',DEFAULT_CONFIG)
```

```python
import copy
class Node:
    def __init__(self,value,next_node=None):
        self.value = value
        self.next = next_node
    def __repr__(self):
        return f"Node({self.value})"
n3 = Node(3)
n2 = Node(2,n3)
n1 = Node(1,n2)
deep_head = copy.deepcopy(n1)
deep_head.next.value = 999
print('Original n2.value:',n2.value)
print('Deep head next value:',deep_head.next.value)
```
