```python
#给你一个 非空 整数数组 nums ，除了某个元素只出现一次以外，其余每个元素均出现两次。找出那个只出现了一次的元素。
#你必须设计并实现线性时间复杂度的算法来解决此问题，且该算法只使用常量额外空间。
#输入：nums = [2,2,1]
#输出：1
#输入：nums = [4,1,2,1,2]
#输出：4
#输入：nums = [1]
#输出：1
from typing import List
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        ans = 0
        for num in nums:
            ans ^= num
        return ans
s = Solution()
print(s.singleNumber([4,1,2,1,2]))
```

```python
#颠倒给定的 32 位有符号整数的二进制位。
#输入：n = 43261596
#输出：964176192
#输入：n = 2147483644
#输出：1073741822
m0 = 0x55555555  # 01010101 ...
m1 = 0x33333333  # 00110011 ...
m2 = 0x0f0f0f0f  # 00001111 ...
m3 = 0x00ff00ff  # 00000000 11111111 00000000 11111111
m4 = 0x0000ffff  # 00000000 00000000 11111111 11111111
class Solution:
    def reverseBits(self, n: int) -> int:
        n = n>>1&m0 | (n&m0)<<1
        n = n>>2&m1 | (n&m1)<<2
        n = n>>4&m2 | (n&m2)<<4
        n = n>>8&m3 | (n&m3)<<8
        return n>>16 | (n&m4)<<16
s = Solution()
print(s.reverseBits(2147483644))
```
